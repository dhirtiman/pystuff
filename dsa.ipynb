```python
def countDigitOcc(d,n):
    occ = 0
    while(n>0):
        occ+= (n%10==d)
        n//=10
    return occ

print(countDigitOcc(1,123123121323123123))
```

```python
def apendNum(n,m):
    x = m
    while(m>0):
        n*=10
        m//=10
    return n+x


print(apendNum(12345,999))
```

```python
def prependNum(n,m):
    x = n
    while(n>0):
        m*=10
        n//=10
    return m+x


print(prependNum(12345,999))
```

```python
def replaceLastPart(n, m):
    x = m
    tens = 1
    while (m > 0):
        n //= 10
        tens *= 10
        m //= 10

    return n*tens + x


print(replaceLastPart(1234, 50))
```

```python
def getLargestAndSmallest(n):
    
    lg = 0
    sm = 9
    while(n>0):
        dig = n%10
        if(dig>lg):
            lg = dig
        if(dig<sm):
            sm = dig
        n//=10
    return lg,sm

print(getLargestAndSmallest(123490))
```
