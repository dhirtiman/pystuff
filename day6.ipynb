```python
def getSign(n):
    if (n < 0):
        return -1
    return 1

def getMagnitude(n):
    if (n < 0):
        return -n
    return n

def getRev(n):
    num = getMagnitude(n)
    rev = 0
    while (num > 0):
        rev = rev*10+(num % 10)
        num = num//10

    return rev*getSign(n)

def isPalindrome(n):
    if n == getRev(n):
        return True
    return False


def getNthPallindromeFromM(n,m):
    
    while(n>0):
        if(isPalindrome(m)):
            n -=1
        m = m+1
    return m-1

print(getNthPallindromeFromM(15,519))
```
