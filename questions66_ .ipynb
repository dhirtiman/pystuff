```python
# 66. Write a program to Calculate Reverse a number using Recursion.


def getRevRec(n, rev=0):
    if (n == 0):
        return rev
    rev = (rev * 10) + n % 10
    return getRevRec(n//10, rev)

def getRev(n):
    if(n<0):
        return -getRevRec(abs(n))
    return getRevRec(n)


print(getRev(123))
print(getRev(-12))
```

```python
# 76 Write a program to check whether a given number is an ugly number.
# In number system, ugly numbers are positive numbers whose only prime factors are 2,
# 3 or 5. First 10 ugly numbers are 1, 2, 3, 4, 5, 6, 8, 9, 10, 12. By convention, 1 is
# included.


# To check if a number is ugly, repeatedly divide by 2, 3, and 5 until it is no longer divisible. If the final number is 1, it is an ugly number.

def isUgly(n):
    divisors =[2,3,5]
    
    for i in divisors:
        if(n%i==0):
            while(n%i==0):
                n = n // i
    if(n-1==0):
        return True
    return False
    


print(isUgly(12))
print(isUgly(15))
print(isUgly(1))
print(isUgly(2))

```

```python
def factorial(n):
    if(n<=0):
        0
    fact = 1
    for i in range(1,n+1):
        fact*=i
    return fact


def getBinomialCoficient(n,r):
    
    if(r<0 or r<0):
        return 0
    if(n==r or r==0):
        return 1
    
    nFact = factorial(n)
    rFact = factorial(r)
    nrFact = factorial(n-r)
    
    return nFact// (rFact*nrFact)

       
def getCatalanNum(n):
    nums = [1]
    for i in range(1,n):
        catalan = getBinomialCoficient(2*i,i) // (i+1)
        nums.append(catalan)
    return nums       
getCatalanNum(10)
```

```python
# 88 Write a program to check whether a number is a Pronic Number or Heteromecic Number or not.

def getMagnitude(n):
    if (n < 0):
        return -n
    return n

def squareRoot(n):
    if n < 0:
        return 0
    if n == 0:
        return 0
    root = n / 2.0
    while 1:
        newroot = 0.5 * (root+(n/root))
        if getMagnitude(root - newroot) < 0.0000000001:
            return newroot
        root = newroot
        

def isPronic(n):
    root = squareRoot(n)
    root = root//1
    prod = root*(root+1)
    if(prod==n):
        return True
    return False
def isHeteromecic(n):
    return isPronic(n)

print(isPronic(110))
print(isPronic(20))
print(isHeteromecic(30))
print(isHeteromecic(12))
```

```python
# 92 write a program to check a number is a cube or not.


def getPrimeFactors(n):    
    factors = []
    for i in range(2,n+1):
        if(n%i==0):
            while(n%i==0):
                n = n//i
                factors.append(i)
    
    return factors

def isPerfectCube(n):
    factors = getPrimeFactors(n)
    for factor in factors:
        count = 0
        for fac in factors:
            if(factor == fac):
                count +=1                
        if(count<3):
            return False
    return True


print(isPerfectCube(375)) 
print(isPerfectCube(8))   
print(isPerfectCube(1000))   
print(isPerfectCube(72))   
```
