```python
# 5. check if the number is prime, if it prime replace
#    the number with 1, else with 0

def isPrime(n):
    if(n<2):
        return False
    if(n<4):
        return True
    if (n%2==0 or n%3==0):
        return False
    i=5
    while i*i <=n:
        if(n%i==0 or n% (i+2)==0):
            return False
        i+=6
    return True


def filterPrimeArr(arr):
    for i in range(len(arr)):
        
        if(isPrime(arr[i])):
            arr[i]=1
        else:
            arr[i]=0
            
    return arr

arr=[65,90,97,122,48,57,32]
print(isPrime(10))
print(arr)
filterPrimeArr(arr)
print(arr)
```

```python
# 8. create 2 new arrays store the the quotient(//) of each elemnt
#    when it is divided with len of the array and reminder(%) of each elem with len
#    in another array

def divideLenArr(arr):

    quotients = []
    remainders = []
    n = len(arr)
    for el in arr:
        remainders.append(el % n)
        quotients.append(el//n)
    return quotients, remainders

arr=[65,90,97,122,48,57,32]

quotients,remainders = divideLenArr(arr)

print("Array:",arr)
print("Quotients:",quotients)
print("Remainders:",remainders)
```

```python
# 9. swap the adj elem
# arr=[65,90,97,122,48,57,32]
# -> 90,65,122,97,57,48,32


def swapAdjacent(arr):

    n = len(arr)

    i = 0
    while i < n-1:
        
        arr[i] = arr[i]^arr[i+1]
        arr[i+1] = arr[i]^arr[i+1]
        arr[i] = arr[i]^arr[i+1]
             
        i += 2
    return arr


arr = [65, 90, 97, 122, 48, 57, 32]
print("Array:",arr)
swapAdjacent(arr)
print("Swapped:",arr)
```

```python
# 11. 1st and last elm as a pair
    # in--> 65,90,97,122,48,57,32
    # out-> 65,32,90,57,97,48,122


def pair1stLstArr(arr):
    pairArr = []
    n = len(arr)
    i=0
    j=n-1
    while (i<=j):
        pairArr.append(arr[i])
        if(j==i):
            break
        pairArr.append(arr[j])
        i+=1
        j-=1
    return pairArr

arr = [65, 90, 97, 122, 48, 57, 32]
print(arr)
arr = pair1stLstArr(arr)
print(arr)
```

```python
# 12. even and odd and at 1 side
#     in--> 65,90,97,122,48,57,32
#     out-> 90,122,48,32,65,97,57
#     out-> 65,97,57,90,122,48,32

def evenOddArrs(arr):
    evens = []
    odds = []

    n = len(arr)

    for i in range(n):
        if (arr[i] % 2 == 0):
            evens.append(arr[i])
        else:
            odds.append(arr[i])

    evenSide = evens+odds
    oddSide = odds+evens

    return evenSide, oddSide


arr = [65, 90, 97, 122, 48, 57, 32]
print("Array:", arr)
evenSide, oddSide = evenOddArrs(arr)

print("Even sided Array:", evenSide)
print("Odd sided Array:", oddSide)
```

```python
# 13. print 3rd max and 3rd min in a array


def sortArr(arr):
    n = len(arr)
    for i in range(n):
        
        for j in range(i+1,n):
            if(arr[i]>arr[j]):
                temp = arr[i]
                arr[i] = arr[j]
                arr[j] = temp
    
    return arr


def getNthMax(arr,nth):
    sortArr(arr)
    n = len(arr)
    if(nth>n):
        return 
    return  arr[n-nth]

def getNthMin(arr,nth):
    sortArr(arr)
    n = len(arr)
    if(nth>n):
        return 
    return  arr[nth-1]    

    
    
    
arr = [65, 90, 97, 122, 48, 57, 32]
print("Array:", arr)
print("3rd Max: ",getNthMax(arr,3))
print("3rd Min: ",getNthMin(arr,3))
```
