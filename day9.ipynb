# Array

```python
arr = [1,2,4]

print(arr)

for el in arr:
    print(el,end=" ")
    
print()
for i in range(len(arr)):
    print(arr[i],end=" ")
```

```python
# maximum element in the given array


def max(arr):
    max=0
    for i in arr:
        if(max<i):
            max=i
    return max

arr = [1,2,3,4,5]

print(arr)
max(arr)
```

```python
# minimum element in the given array


def min(arr):
    min=arr[0]
    for i in arr:
        if(min>i):
            min=i
    return min

arr = [1,2,3,4,5,-1]

print(arr)
min(arr)
```

```python
# nth minimum element in the given array


def min(arr,n):
    min=arr[0]
    wo = []
    for i in arr:
        if(min>i):
            wo.append(min)
            min=i
    return wo

arr = [1,2,3,4,5,-1]
print(arr)
min(arr,1)
```

```python
#write a method to print 2nd maximum and 2nd minimum from a given array

def min2nd(arr):
    min = arr[0]
    min2 = arr[0]
    for i in arr:
        if(min>i):
            min2=min
            min=i
        elif(min2>i and min!=i):
            min2=i
    return min2

def min2nd(arr):
    
    # mins = []
    
    mins = [arr[0],arr[0]]
    for i in arr:
        if(mins[0]>i):
            mins[1]=mins[0]
            mins[0]=i
        elif(mins[1]>i and mins[0]!=i):
            mins[0]=i
    return mins


def max2nd(arr):
    max = 0
    max2 = 0
    
    for i in arr:
        max-=i
    max2=max
    for i in arr:
        if(i>max):
            max2=max
            max=i
        elif(i>max2 and max!=i):
            max2=i
    return max2



def max2nd2(arr):
    max = arr[0]
    max2 = arr[0]
    for i in arr:
        if(i>max):
            max2=max
            max=i
        elif(max2<i or max==max2 ):
            max2=i
    return max2



# arr = [1,2,3,4,5,-1]

# print(min2nd(arr))
# print(max2nd(arr))

arr = [5,4,3,2,1]
arr = [1,2,3]
print(max2nd(arr))
print(max2nd2(arr))
            
```

```python

def min2nd(arr):
    min = arr[0]
    min2 = arr[0]
    for i in arr:
        if(min>i):
            min2=min
            min=i
        # elif(min2>i and min!=i):
        elif(min2>i or min==min2):
            min2=i
    return min2

arr = [1,2,3,4,5]
print(min2nd(arr))
```

```python
# write a method to add all the number of a given array


def arrSum(arr):
    sum=0
    for el in arr:
        sum+=el
    return sum

arr = [1,2,3,4,5]
arrSum(arr)
```

```python
# find element , return index , if not found return -1


def findEl(arr,el):
    for i in range(len(arr)):
        if(el==arr[i]):
            return i
    
    return -1

arr = [1,2,3,4,5]
print(findEl(arr,3))
print(findEl(arr,6))
```

```python
def findEvenAndOddElem(arr):
    oddCount=0
    evenCount=0
    for i in range(len(arr)):
        if(arr[i]%2==0):
            oddCount=oddCount+1
        else:
            evenCount=evenCount+1
    return oddCount,evenCount


arr=[10,12,13,17,23]


print(findEvenAndOddElem(arr))
```

```python
# 1


def numLen(num):
    len=0
    while(num>0):
        num=num//10
        len+=1
    return len

def replaceElLen(arr):
    
    for i in range(len(arr)):
        arr[i] = numLen(arr[i])
    
    return arr

arr=[12,7,321,11,7080]

print(replaceElLen(arr))

    
```

```python
def isPrime(n):
    if(n<2):
        return False
    if(n<4):
        return True
    if (n%2==0 or n%3==0):
        return False
    i=5
    while i*i <=n:
        if(n%i==0 or n% (i+2)==0):
            return False
        i+=6
    return True


def filterPrimeArr(arr):
    for i in range(len(arr)):
        
        if(isPrime(arr[i])):
            arr[i]=1
        else:
            arr[i]=0
            
    return arr

arr=[12,7,321,11,7080]
print(filterPrimeArr(arr))
arr=[1,2,3,4,5,6,7,8,9]
print(filterPrimeArr(arr))
```

```python
def toBin(n):
    tens = 1
    num = 0
    while (n > 0):
        rem = n % 2
        num = num + rem*tens
        tens *= 10
        n = n//2
    return num

def toBinArr(arr):
    for i in range(len(arr)):
        arr[i] = toBin(arr[i])
        
    return arr

arr = [11,7,15,64,100]

toBinArr(arr)
print(arr)
```

```python

def revArr(arr):
    n = len(arr)
    i=0
    j=n-1
    
    while (i<j):
        
        arr[i] = arr[i]^arr[j]
        arr[j] = arr[i]^arr[j]
        arr[i] = arr[i]^arr[j]
                
        i+=1
        j-=1
    return arr



arr = [1,2,3,4,5,6]


print(revArr(arr))
```

```python
def rotateArr(arr,k):
    n = len(arr)
    while k > n:
        k-=n
    for _ in range(k):
        for i in range(n-1,0,-1):
            # print((i+1)%n)
            pos = (i+1)%n
            tem = arr[pos]
            arr[pos]=arr[i]
            arr[i] = tem
                
    return arr    
    



arr = [8,2,0,4,1,4,2,1,0,6,6,2,5,6,6,2,7,9,4,1,3,9,6,5,4,8,7,8,9,2,5,5,8,3,0,5,2,5,3,9,8,5,8,8,6,3,0,2,8,1,8,4,6,4,1,6,4,3,7,9,3,0,3,9,3,3,2,1,3,2,8,7,7,7,2,0,3,1,2,1,7,7,2,8,4,0,4,3,1,9,1,5,9,8,5,6,4,2,8,0,9,6,5,7,2,6,3,1,2,1,0,6,9,7,5,3,9,8,2,6,1,8,6,6,4,4,7,3,3,5,3,2,2,9,2,7,5,2,8,5,8,7,5,3,6,0,4,1,0,8,9,0,1,2,6,0,0,3,4,1,6,6,5,9,2,5,6,7,8,4,4,5,0,8,1,1,7,9,5,2,0,1,6,2,6,1,1,3,6,5,8,7,3,8,9,6,0,0,8,9,4,0,1,6,7,8,3,9,5,1,4,6,7,3,4,7,6,3,0,1,3,9,3,1,6,4,8,8,3,8,4,7,6,7,3,4,0,1,7,6,2,5,5,2,9,9,0,9,5,9,8,3,8,3,7,9,1,9,4,0,7,6,9,0,6,8,7,9,5,5,0,7,8,8,3,4,3,8,2,6,5,8,1,3,9,0,7,6,1,4,3,7,9,3,9,3,8,8,6,8,1,5,8,2,5,2,1,2,4,6,6,4,8,7,0,8,6,1,0,9,2,3,6,7,4,8,2,0,0,0,7,3,5,4,6,7,0,0,0,1,9,0,2,7,1,1,4,5,3,7,1,2,0,9,6,6,3,4,5,8,8,4,0,3,8,3,0,4,3,5,4,7,8,6,8,2,6,1,1,6,9,0,4,5,2,1,1,1,3,5,3,8,2,6,2,4,9,4,0,7,5,2,7,4,9,6,8,8,5,7,1,7,8,1,7,0,1,6,4,3,9,1,7,4,4,0,1,0,8,9,3,7,3,3,4,9,7,7,4,9,1,8,7,9,0,0,2,3,8,9,1,0,2,6,7,0,5,6,4,5,7,4,9,4,7,3,3,2,0,4,7,4,7,2,3,7,1,6,3,7,8,1,5,4,3,2,9,6,8,0,7,4,8,3,7,7,2,6,0,1,4,4,9,0,1,1,6,8,9,5,0,2,0,5,5,8,5,1,3,6,8,9,5,7,0,0,7,2,5,6,9,6,6,3,6,3,7,8,5,3,5,9,1,4,1,1,1,5,1,4,0,0,4,9,3,3,9,5,1,4,1,8,7,9,9,2,4,9,2,9,5,2,8,0,6,5,9,0,0,6,6,8,8,3,9,3,1,6,9,4,3,7,8,0,4,2,8,6,7,8,2,1,5,7,4,9,9,7,1,7,1,1,4,8,3,4,7,8,2,5,5,4,6,9,3,2,7,2,6,1,4,2,5,8,3,6,4,4,9,4,0,6,8,4,3,6,8,5,1,0,3,5,2,3,2,9,1,6,4,8,3,3,2,7,0,7,7,8,8,5,3,0,6,8,5,8,8,0,9,9,2,1,2,3,1,2,7,5,4,5,6,9,6,0,8,9,9,8,7,3,4,1,8,7,7,0,7,3,6,3,0,8,0,4,1,8,1,4,8,1,5,4,9,4,4,5,1,5,8,7,6,8,5,8,4,4,1,5,3,9,4,8,6,8,6,3,4,8,7,0,6,8,1,8,9,8,1,9,1,4,9,2,8,2,6,7,1,9,1,0,3,6,8,3,5,4,9,3,6,1,2,6,8,7,2,3,3,3,3,2,3,9,2,4,6,1,5,7,3,8,4,6,9,9,5,0,2,1,0,6,1,9,6,7,9,6,6,7,0,3,1,9,2,4,9,3,8,3,7,3,1,9,4,4,0,3,5,9,4,5,0,2,3,4,5,9,1,0,6,5,5,7,5,4,0,9,8,2,0,7,8,7,6,4,8,6,8,0,7,1,3,9,7,7,0,9,8,5,3,9,8,2,7,2,0,8,9,6,4,8,4,4,0,6,5,8,6,0,0,9,8,6,4,7,9,3,3,2,7,9,1,9,3,2,3,7,9,5,7,3,8,7,5,5,5,1,3,7,4,1,4,9,4,3,5,1,6,8,0,7,3,1,8,3,4,5,4,5,2,7,0,9,0,9,8,0,4,0,0,7,9,8,7,4,9,0,7,9,9,7,9,7,0,2,6,2,0,9,9,4,9,5,9,7,7,6,8,9,1,6,5,9,7,0,5,0,1,2,3,7,0,5,6,4,0,3,7,9,1,8,0,3,6,2,1,1,8,8,4,9,5,5,2,1,7,5,0,8,7,0,3,4,4,5,7,2,0,4,4,8,9,5,4,0,8,5,3,4,0,5,8,0,0,2,4,1,4,3,4,6,6,9,0,8,4,2,7,7,9,4,2,1,5,1,7,5,5,7,4,1,7,5,7,6,6,5,2,6,7,1,6,9,2,9,5,1,3,6,0,1,5,9,6,5,3,8,3,9,9,2,6,8,6,3,0,9,4,6,7,8,2,8,5,9,3,6,3,5,9,0,1,5,5,9,2,5,7,1,8,2,5,1,8,0,0,1,3,1,4,1,8,2,6,9,3,9,4,4,7,4,9,1,5,0,9,0,5,5,1,1,1,3,2,6,2,2,9,1,7,7,4,1,3,1,0,7,8,5,1,2,7,4,2,6,3,5,3,6,2,4,1,6,3,6,7,4,2,0,4,6,7,0,1,3,5,0,1,4,8,3,1,9,2,0,0,1,9,8,5,7,0,5,6,1,6,2,9,9,8,5,6,1,5,1,1,8,8,5,2,6,2,0,8,0,1,0,8,0,9,5,7,8,7,6,6,6,0,4,2,4,1,5,8,3,6,2,0,4,0,8,3,9,3,5,0,5,3,1,4,1,4,8,5,3,7,9,3,0,7,3,4,5,4,6,4,4,7,6,3,0,2,8,1,7,8,5,6,1,5,7,1,8,1,5,0,7,6,4,4,6,2,1,7,1,7,9,3,0,1,6,9,9,5,2,5,3,8,3,8,6,4,3,2,1,5,5,2,0,8,2,0,9,6,9,7,4,1,9,2,6,0,8,1,4,9,0,9,5,8,5,4,6,3,8,5,3,0,5,4,5,6,7,1,9,2,8,5,8,6,8,6,4,7,1,0,0,2,2,0,3,9,1,4,6,6,1,0,7,2,3,1,2,8,3,6,5,5,4,5,0,2,1,7,6,1,6,2,5,0,1,5,3,0,8,8,9,5,8,2,9,9,1,7,4,5,1,3,3,8,0,7,4,2,6,1,4,9,5,3,6,6,6,9,5,6,4,0,6,0,3,0,9,0,3,9,3,6,1,0,5,6,9,8,6,5,9,8,2,2,2,1,4,9,2,7,0,9,2,4,9,8,7,5,3,8,8,2,2,0,3,5,6,4,7,9,5,8,4,1,6,4,1,6,6,4,3,9,5,3,9,5,0,4,5,8,4,5,8,4,7,9,8,0,5,9,8,6,8,9,6,0,9,6,6,7,6,5,8,8,2,3,5,7,3,1,1,3,0,2,7,8,5,6,3,7,5,1,0,0,3,6,2,8,5,7,2,8,4,1,6,8,6,6,1,5,6,0,2,1,1,5,7,8,7,5,1,9,8,7,5,3,9,6,4,1,7,3,3,7,6,9,0,5,3,2,4,4,6,2,0,5,7,0,3,3,6,3,2,2,9,1,6,9,8,3,5,5,1,3,0,0,1,5,8,4,3,3,5,6,0,6,8,1,6,2,4,9,7,8,1,8,4,3,7,2,8,4,1,7,8,2,7,6,0,8,7,9,7,2,2,2,4,6,9,2,1,8,6,1,1,7,0,4,5,6,0,3,2,2,5,7,6,7,7,7,4,1,7,5,9,7,0,2,8,3,0,7,4,6,8,8,5,4,3,4,2,8,1,1,3,6,9,1,7,4,8,3,7,3,1,9,8,4,6,2,6,7,7,4,4,2,1,1,9,4,8,2,2,3,2,8,7,8,0,2,9,3,1,7,6,4,0,2,3,4,4,2,3,6,0,9,8,9,5,4,2,1,2,1,8,5,7,9,7,3,7,3,3,6,4,9,4,9,0,4,7,9,1,0,3,7,7,4,9,9,6,3,5,4,0,7,7,2,0,8,5,0,0,1,7,1,0,0,0,9,7,0,5,0,2,4,9,2,7,4,5,9,0,6,9,7,7,9,3,3,6,9,2,5,3,2,4,8,1,8,4,1,7,8,0,6,4,3,8,8,4,8,3,1,5,7,4,8,2,2,7,9,1,7,5,9,0,1,5,3,2,7,5,7,1,8,1,2,1,9,0,4,5,6,0,6,1,3,3,3,4,6,8,4,5,4,4,3,0,5,2,0,3,5,0,9,0,4,9,0,7,1,1,1,9,9,9,4,6,1,9,8,9,0,6,1,2,2,0,8,6,6,6,2,4,0,0,5,3,7,7,5,1,2,3,3,5,2,5,5,7,5,2,0,1,6,7,5,4,1,1,4,2,4,9,0,3,6,8,4,8,9,3,0,6,1,0,7,6,2,4,6,7,3,9,2,3,3,7,2,8,5,3,4,1,3,4,3,2,7,8,4,8,1,7,4,0,5,6,6,5,0,3,1,6,6,6,5,7,1,6,1,9,4,9,2,8,6,1,7,9,7,6,6,0,0,1,6,2,6,2,9,3,5,0,7,5,1,6,5,4,8,1,0,0,2,1,1,7,0,1,8,5,6,3,6,0,4,2,2,4,9,5,7,7,1,0,3,4,9,8,1,2,3,4,1,9,0,1,7,3,1,6,2,9,1,9,2,2,4,4,9,8,3,8,2,4,8,4,7,5,1,3,3,6,9,6,4,1,8,3,7,8,0,2,3,3,9,7,5,5,5,1,8,1,9,6,1,3,8,1,0,9,5,6,2,0,4,1,3,2,0,4,1,5,7,3,2,4,8,2,5,1,6,2,0,1,2,3,3,8,3,7,2,5,8,6,4,6,9,0,4,9,1,7,2,0,5,1,9,7,2,9,2,8,5,5,7,6,0,3,1,0,5,0,8,4,0,8,1,4,1,2,6,7,1,3,8,4,9,4,1,0,1,3,6,0,3,0,3,7,9,5,8,2,9,8,2,0,4,0,7,2,6,0,7,1,9,3,5,8,3,6,9,4,7,0,9,4,0,9,2,5,8,6,1,6,2,8,2,5,4,4,5,5,2,2,5,4,4,5,1,2,3,0,0,9,2,5,4,7,0,0,7,2,5,8,0,1,4,8,0,2,2,3,6,9,7,7,3,2,1,8,8,9,7,5,6,4,9,9,2,9,5,7,4,0,1,2,6,6,3,1,8,9,4,2,0,0,8,3,6,2,0,8,7,2,8,3,6,4,8,7,1,6,5,1,4,2,7,6,6,1,9,9,9,1,9,7,6,4,1,5,0,5,1,1,8,2,7,3,7,0,9,6,0,7,5,0,2,5,0,7,4,3,6,9,3,9,3,7,7,9,9,9,1,5,0,3,8,6,2,7,3,3,0,5,5,5,5,7,9,6,3,4,1,9,9,6,6,4,8,7,0,0,6,0,3,1,4,0,5,2,8,4,7,7,1,8,1,4,4,7,3,2,3,5,4,4,0,0,7,4,0,2,6,9,9,0,3,5,9,7,4,6,1,4,8,3,8,9,2,9,0,1,0,5,8,7,4,6,1,3,2,7,4,8,3,5,2,6,2,7,4,7,9,3,2,8,0,2,8,3,3,3,3,8,6,5,2,7,6,7,9,0,7,7,3,5,3,6,2,6,1,6,9,8,6,3,2,5,3,1,9,0,8,4,6,7,8,7,2,5,6,4,9,5,1,7,5,6,2,8,0,9,1,1,2,8,6,5,7,6,4,5,3,5,4,6,9,5,4,8,8,5,6,9,7,2,9,2,1,1,6,5,2,1,5,3,0,6,1,4,2,4,4,4,9,6,5,8,7,0,6,8,7,4,7,5,6,1,4,5,2,1,8,4,7,1,3,3,3,6,7,3,2,9,4,6,5,4,3,2,7,0,2,9,1,7,6,0,5,3,5,8,7,8,7,1,5,1,2,6,8,3,6,1,8,9,5,9,7,4,3,2,7,7,8,3,3,0,1,8,0,3,5,4,0,4,5,6,1,1,6,6,1,3,0,4,7,4,5,4,2,3,9,5,0,9,7,0,7,6,8,9,3,0,1,9,8,2,3,7,0,8,9,5,7,2,0,1,6,6,9,7,1,0,8,4,1,0,5,0,3,5,6,1,7,2,9,0,6,5,2,1,1,8,3,2,5,1,8,0,8,2,2,7,6,7,8,0,2,2,5,9,9,9,3,4,6,6,0,7,6,8,7,5,1,9,0,6,3,9,0,0,9,1,1,3,7,5,1,0,7,7,9,7,2,0,3,9,8,0,0,3,2,1,6,1,5,4,0,5,7,5,7,6,4,1,9,1,4,1,0,8,2,0,3,8,7,7,0,9,7,7,7,6,8,2,1,9,1,2,9,2,6,8,7,2,6,6,8,7,2,6,8,6,1,9,4,4,6,5,2,1,0,2,2,2,4,3,0,1,4,6,0,0,6,4,5,8,4,3,5,3,4,8,8,7,6,3,1,4,5,2,6,5,1,8,4,8,6,1,4,3,1,2,5,3,1,7,9,9,7,4,7,5,7,6,9,1,0,4,6,3,3,4,6,4,8,2,8,9,1,1,3,0,0,0,0,5,1,8,0,4,6,2,6,2,8,1,6,4,9,0,5,0,4,2,6,4,4,9,2,6,9,2,8,1,8,0,3,0,3,7,2,5,1,7,8,5,6,2,0,4,8,1,2,8,3,8,8,6,6,2,2,4,3,5,7,6,1,2,3,6,2,8,2,8,9,7,4,8,1,4,3,2,5,3,2,5,8,7,7,3,8,6,9,9,3,1,0,3,5,2,9,2,0,2,1,4,3,1,6,9,8,2,6,5,2,0,9,7,1,7,8,9,6,9,4,4,9,5,5,6,0,4,4,6,1,0,7,0,6,2,1,6,4,3,4,6,9,1,9,8,3,8,3,1,9,3,1,4,4,3,9,3,4,9,7,9,4,1,9,8,9,1,1,8,2,2,7,1,7,3,5,9,4,8,7,0,9,8,1,2,6,7,1,0,6,2,4,3,8,3,7,1,6,2,3,9,5,4,0,3,1,5,1,4,5,8,8,6,5,2,5,5,9,6,7,6,4,2,6,8,7,1,6,5,5,6,2,2,4,4,8,3,9,4,9,7,4,4,4,5,5,2,5,0,7,9,6,0,2,7,4,7,2,8,4,3,2,9,4,0,3,2,4,7,4,0,2,0,2,6,1,5,5,1,1,9,7,6,9,0,2,9,9,0,7,4,8,5,4,4,7,0,5,2,2,5,3,0,9,9,9,8,7,0,0,0,3,0,1,3,5,9,5,5,9,6,7,4,3,0,7,8,3,7,4,1,7,8,2,5,4,3,3,7,3,5,8,5,1,9,2,8,6,4,3,6,5,7,9,8,4,2,9,3,5,6,8,5,7,6,0,4,4,3,7,1,7,6,7,2,5,1,0,1,3,6,4,3,1,3,1,5,9,8,2,8,2,8,7,3,0,5,9,5,0,6,4,9,9,9,5,2,2,6,8,9,0,6,5,4,5,2,1,8,8,5,6,1,7,9,0,2,4,1,9,5,3,9,4,8,1,5,6,6,5,3,9,9,8,3,2,0,0,3,5,5,6,3,3,5,3,9,1,3,5,1,8,4,9,2,1,0,0,9,5,6,3,6,4,2,0,0,7,3,9,2,0,2,0,4,7,3,2,9,0,9,7,5,4,7,8,4,5,3,5,2,5,7,1,0,5,3,8,8,9,0,7,3,1,6,4,4,8,5,9,1,2,7,1,1,1,7,3,1,3,9,9,3,6,9,2,7,6,2,5,1,0,0,2,3,0,7,5,0,5,9,6,3,1,9,2,4,6,7,1,4,9,5,8,7,1,7,7,2,0,3,8,8,4,6,1,3,5,7,0,6,3,2,3,0,2,5,2,3,2,6,9,0,5,7,0,2,1,0,8,9,8,4,4,5,0,5,1,0,7,4,2,7,2,8,4,8,5,6,9,0,8,1,1,1,8,7,6,3,9,4,5,3,6,3,7,8,4,2,0,5,2,1,5,3,7,8,0,5,0,4,6,2,1,4,0,2,8,1,9,3,2,9,0,9,0,6,1,8,7,4,9,3,8,6,8,0,2,6,1,9,9,5,2,0,1,0,6,7,2,5,5,7,3,5,1,1,0,4,3,2,3,1,7,7,3,4,4,8,1,5,6,8,6,2,1,1,8,5,1,1,4,5,0,3,5,0,6,8,5,7,6,6,1,5,1,2,7,6,1,0,1,5,9,7,2,0,8,6,8,2,3,2,9,9,2,6,8,6,7,9,5,9,4,0,9,2,7,1,5,5,4,2,6,6,6,6,7,6,0,2,9,4,2,0,4,3,1,6,1,1,2,2,5,7,1,0,1,2,8,8,7,1,7,7,9,9,7,3,6,6,7,4,4,4,4,6,7,6,3,3,2,4,5,9,8,1,3,3,3,1,0,3,2,1,7,7,3,4,9,4,9,6,7,3,4,2,6,7,8,7,5,2,6,7,7,4,0,4,9,4,7,4,6,0,0,9,1,7,4,5,6,4,3,1,9,3,3,5,7,1,0,8,4,5,3,1,1,8,6,3,8,9,4,1,5,1,4,7,4,8,7,8,4,9,2,9,1,8,3,0,6,7,5,8,8,4,6,7,9,4,4,3,5,0,7,1,9,4,8,1,1,4,4,4,9,7,8,5,8,7,3,9,1,7,4,6,8,8,3,1,6,9,4,6,7,1,9,6,4,5,6,5,5,3,6,6,9,2,0,2,6,2,7,6,9,3,1,1,3,1,3,8,5,3,3,1,7,6,9,6,2,2,8,4,6,9,3,3,7,9,4,4,5,1,7,9,8,6,4,6,4,2,6,2,6,1,8,8,2,0,7,4,4,9,9,3,4,8,0,1,7,3,3,7,8,5,3,6,4,8,0,7,4,1,4,6,1,8,4,4,3,2,5,3,8,1,8,5,2,7,5,1,5,3,1,9,6,6,6,2,1,4,7,3,9,2,4,4,1,5,3,3,7,3,2,6,1,5,0,1,1,4,1,9,2,9,1,8,0,7,1,1,1,0,7,1,1,5,7,5,1,9,0,6,5,4,7,9,8,6,3,4,5,2,0,1,1,8,1,5,1,4,9,8,4,1,2,1,2,2,0,9,9,5,5,9,5,8,3,6,2,4,8,4,6,8,1,6,1,5,7,3,9,8,0,2,4,4,6,9,8,1,9,7,2,8,1,4,6,1,0,0,1,2,8,4,7,7,2,8,0,5,8,1,9,7,5,6,2,2,0,3,3,1,0,0,6,0,5,3,4,9,8,3,0,0,6,5,8,0,2,8,5,7,8,5,3,3,4,1,8,5,9,9,5,8,7,7,5,2,5,4,6,6,4,4,6,6,0,5,4,6,8,4,0,4,0,5,4,3,8,5,0,9,5,2,6,4,4,9,9,4,2,7,4,5,9,6,5,4,1,0,8,6,7,7,0,1,1,5,4,9,9,2,5,8,4,3,3,9,9,0,6,7,6,1,2,9,1,1,9,2,7,1,6,2,4,4,4,2,2,3,4,6,1,9,6,7,0,8,1,8,8,8,2,0,7,5,5,1,9,1,0,0,7,9,5,8,2,3,9,5,5,1,6,9,4,6,1,2,5,1,4,3,3,7,0,6,8,1,9,0,2,9,9,9,7,4,3,7,1,4,7,0,4,7,4,6,3,3,5,8,1,1,7,2,6,2,0,4,6,7,9,4,4,1,8,0,9,7,4,2,6,5,4,1,0,8,1,6,4,7,0,8,1,1,0,9,4,0,0,7,5,9,4,4,6,4,7,6,6,5,2,2,4,5,6,8,1,4,3,1,5,5,0,8,2,5,9,3,8,7,2,8,7,5,1,6,1,0,8,4,1,9,1,1,0,5,0,2,4,6,1,3,9,4,8,1,1,6,4,1,7,6,8,5,5,3,2,9,6,1,6,2,7,6,5,2,7,8,2,5,6,1,9,6,6,4,6,4,4,4,8,0,8,3,4,1,1,9,7,5,4,1,4,9,5,1,5,6,7,6,8,4,6,1,5,7,3,9,6,2,4,3,1,7,7,8,7,5,2,5,3,7,9,4,7,4,0,0,7,6,3,1,0,8,5,2,6,5,0,4,2,6,2,1,3,9,6,9,5,3,3,8,9,3,5,7,7,2,9,4,7,6,3,8,7,1,3,4,9,2,7,9,1,4,4,3,8,6,3,2,8,2,6,3,2,3,3,5,7,7,3,1,8,2,3,3,6,0,2,3,6,7,7,3,8,9,2,6,9,3,2,5,0,5,1,0,1,7,0,6,3,7,0,3,9,1,3,0,4,7,3,6,7,8,3,2,3,5,5,9,9,4,6,5,1,4,9,6,4,7,7,8,7,5,4,5,0,2,2,3,4,3,2,1,5,9,5,1,6,0,6,3,1,0,1,9,4,8,5,2,3,3,7,5,9,6,9,3,8,8,9,5,8,0,4,8,6,6,0,4,3,8,2,9,5,3,2,5,5,3,7,6,3,0,7,5,1,1,7,0,8,1,9,2,3,1,0,0,7,7,0,6,4,0,8,0,2,3,2,1,7,9,3,5,4,8,4,7,0,4,5,6,6,1,2,9,2,8,0,3,9,8,5,8,2,7,4,1,2,5,4,2,7,6,4,7,5,4,7,4,3,1,0,2,1,3,8,3,3,7,7,1,1,9,9,0,7,0,6,6,5,5,4,2,7,5,5,3,3,9,1,2,5,2,6,8,1,5,4,3,8,9,5,2,7,2,8,5,8,2,8,4,5,7,1,0,0,1,6,1,4,5,6,6,5,0,0,9,6,2,0,4,8,3,9,7,4,3,6,1,2,5,8,4,4,4,2,6,4,7,2,5,0,4,4,8,7,2,0,1,9,4,9,6,0,4,8,7,1,8,8,7,1,0,9,0,8,9,5,8,5,9,4,5,3,5,2,0,8,5,9,3,8,6,9,6,5,9,7,1,3,8,4,8,2,3,8,0,8,3,3,3,5,9,4,4,7,8,1,4,0,4,1,8,4,0,9,5,7,9,3,4,9,0,1,6,3,0,6,7,0,5,7,8,5,9,5,1,9,6,0,5,1,6,4,0,1,6,1,9,0,3,9,3,3,2,8,6,3,6,3,5,5,1,0,4,8,9,6,3,5,3,0,9,7,4,3,3,9,0,5,8,4,0,5,1,6,3,9,7,6,1,5,0,9,0,7,2,5,1,8,3,2,5,1,4,6,7,2,4,2,2,4,1,2,9,8,8,6,5,2,4,6,9,4,6,4,3,1,8,8,3,8,8,2,4,6,3,0,2,7,3,7,7,1,0,7,7,8,7,9,8,0,8,4,1,4,0,6,6,7,7,0,8,1,5,2,5,4,7,8,0,9,0,6,8,8,4,0,3,8,4,4,2,0,0,1,3,1,8,5,7,5,9,2,3,4,4,7,8,3,1,3,4,1,7,1,3,7,3,7,7,9,6,6,1,5,5,3,6,8,1,7,5,6,2,7,6,1,0,3,8,4,7,8,5,0,1,0,3,0,8,1,7,3,9,3,1,1,6,1,9,8,5,1,4,6,2,1,5,4,6,0,1,8,4,9,0,5,4,4,7,1,9,1,3,9,7,8,9,8,5,8,7,3,9,2,5,6,9,9,9,0,8,3,7,3,6,6,7,8,5,2,3,4,9,4,5,6,9,0,4,0,7,0,4,5,0,5,5,6,9,6,8,5,7,6,8,3,6,6,5,9,6,8,0,6,4,3,1,7,3,6,5,1,6,1,7,6,6,3,6,3,8,1,6,1,7,1,2,8,8,6,1,2,2,7,0,1,2,4,2,4,4,5,8,4,8,1,8,0,7,8,9,3,9,5,9,8,2,7,6,2,7,3,5,7,7,1,4,8,5,0,5,1,1,6,8,4,6,2,1,7,5,8,3,3,7,0,7,8,1,2,9,9,6,6,8,8,4,9,7,1,5,8,4,5,4,2,9,4,7,1,0,6,4,9,5,2,8,4,5,5,8,0,9,5,7,9,2,4,5,8,6,4,0,4,1,0,1,7,8,5,2,2,6,4,3,5,4,1,5,6,8,0,5,8,8,8,1,2,8,0,8,9,6,5,8,4,7,3,6,5,0,2,6,0,3,4,6,7,2,6,8,2,7,3,0,4,2,1,1,0,8,2,2,3,2,3,7,2,2,7,3,6,2,2,0,0,4,9,3,1,0,5,0,3,2,6,3,7,1,7,7,8,8,1,5,1,3,0,8,2,0,2,8,2,5,3,5,8,3,5,5,1,5,2,4,7,7,8,0,9,4,8,9,2,1,5,8,7,1,4,8,1,7,7,4,9,4,9,2,1,3,6,1,9,0,0,1,7,4,2,0,5,8,8,6,4,9,7,8,6,9,4,8,4,0,3,9,8,1,0,0,3,2,7,7,7,3,8,6,2,7,8,2,5,8,4,0,1,5,1,4,6,5,8,1,1,1,6,0,7,2,8,5,9,1,2,3,0,3,9,5,1,2,9,7,4,8,0,3,2,7,7,5,7,7,6,6,4,2,7,6,0,6,0,8,1,6,9,3,1,4,4,0,6,7,0,0,9,1,1,4,5,6,0,7,3,4,0,9,2,0,9,0,6,6,9,6,8,6,9,5,6,0,1,7,3,6,1,3,3,9,5,9,6,2,1,9,0,1,9,9,3,2,5,8,7,8,6,3,0,4,8,3,6,7,1,6,8,3,6,9,3,3,0,1,3,6,1,8,1,1,9,0,8,2,9,0,3,4,0,5,8,8,3,4,8,5,5,4,4,4,8,0,9,2,6,8,9,1,0,1,9,7,4,6,6,8,7,1,3,2,1,1,3,9,6,1,6,2,5,6,8,1,9,6,6,9,7,0,3,3,5,7,0,7,1,3,9,5,6,0,3,1,1,6,1,7,9,8,9,7,3,5,3,9,5,6,5,8,8,8,6,2,3,5,4,0,7,5,2,4,8,2,9,2,8,0,5,9,2,8,8,4,0,2,8,2,3,7,6,1,1,3,5,0,0,5,2,5,2,9,6,4,3,0,1,4,1,0,3,6,5,4,1,6,9,6,1,3,7,7,0,3,4,8,6,0,3,1,4,2,7,4,5,2,8,6,9,6,0,6,0,4,4,7,2,1,9,7,7,7,9,7,9,5,1,9,6,1,4,6,4,5,5,5,0,3,2,6,5,0,8,7,8,8,2,8,6,1,1,1,9,5,2,2,3,6,7,4,1,4,8,2,1,4,9,1,3,1,5,7,6,2,9,5,1,7,6,3,5,9,6,9,4,4,8,4,2,2,6,6,5,8,4,5,0,4,6,7,9,4,9,8,4,4,5,4,7,8,0,2,7,1,3,1,9,1,3,4,4,7,6,7,3,5,1,4,6,1,9,8,7,2,5,4,8,7,7,3,2,2,9,2,6,7,0,9,8,1,3,2,1,2,6,8,0,5,6,5,9,3,3,1,4,3,0,3,7,5,3,1,9,9,5,3,2,3,8,4,1,3,2,7,2,0,5,0,1,3,9,2,4,3,4,0,2,9,1,8,0,1,7,2,9,6,6,0,3,5,7,8,6,0,5,3,2,5,2,0,5,3,3,0,8,5,7,6,2,1,8,2,4,9,9,0,0,4,8,1,0,6,1,5,3,9,3,9,6,7,6,7,2,1,7,3,6,6,1,5,7,2,1,2,0,2,7,1,7,4,4,4,2,8,9,1,2,6,0,1,6,4,5,2,0,8,7,5,6,3,6,6,0,6,3,8,6,7,3,9,0,7,4,5,7,4,8,2,2,3,8,0,1,0,8,0,4,8,8,6,3,0,6,5,7,2,0,0,8,8,9,5,3,9,4,7,7,4,7,6,2,2,2,3,2,5,8,1,2,7,0,4,6,1,0,5,9,4,8,7,7,1,2,0,5,6,4,1,5,6,3,0,4,4,0,9,3,3,8,0,4,3,9,0,1,4,1,8,5,9,7,3,8,3,7,5,9,8,0,6,2,6,8,1,5,5,6,7,9,4,9,2,7,0,0,2,1,9,0,4,2,1,0,7,9,5,0,3,1,8,7,3,3,7,3,6,6,0,1,6,4,2,8,9,2,4,1,0,8,8,8,5,8,6,0,9,1,4,3,8,0,8,4,0,1,3,4,4,8,4,1,1,2,5,8,1,1,3,2,1,2,9,4,8,6,0,9,7,0,9,0,1,6,0,6,3,9,1,3,4,9,5,1,7,6,8,5,8,5,8,9,5,2,9,6,5,7,7,8,9,6,0,9,7,0,1,8,9,8,5,2,4,3,8,9,9,2,9,1,5,6,3,4,2,0,7,0,8,5,8,5,9,1,2,5,3,9,2,6,9,9,2,7,6,2,7,9,2,0,2,4,6,4,3,2,9,4,8,4,1,9,3,9,8,4,0,5,2,5,2,0,5,8,0,2,1,9,9,1,8,8,9,1,9,1,2,3,0,3,2,0,4,7,7,1,9,7,5,7,1,5,5,5,9,1,2,6,0,0,3,4,6,8,1,2,9,4,1,5,2,1,7,6,4,4,2,1,4,9,0,8,9,0,7,2,5,6,7,4,9,5,8,0,4,8,1,0,0,6,1,4,5,0,2,5,1,3,1,8,2,0,6,6,4,0,0,1,0,6,7,3,0,9,1,7,9,3,3,9,7,8,9,1,2,2,2,7,0,9,5,9,4,5,7,5,9,3,5,4,4,6,3,8,8,9,2,5,6,3,1,4,3,8,4,1,6,6,7,4,8,8,5,2,8,8,1,2,1,7,6,8,8,6,3,5,8,3,2,1,2,3,8,1,1,1,3,0,6,4,3,9,4,9,1,1,9,8,0,2,0,6,5,8,0,5,7,0,3,8,5,6,7,9,3,7,1,1,7,4,2,6,1,1,3,7,3,7,2,2,5,1,9,8,6,2,4,4,4,1,6,5,8,4,4,1,8,6,1,4,1,7,2,9,8,3,5,3,0,8,7,9,3,4,0,0,5,9,5,4,9,5,0,4,2,3,3,4,9,1,1,1,6,7,8,6,8,8,3,5,7,2,3,4,3,4,9,3,4,5,8,7,9,1,0,4,0,2,1,9,1,0,2,8,6,0,0,6,1,3,7,4,3,9,9,5,4,5,7,3,2,4,9,6,3,3,8,2,6,3,2,3,2,8,4,4,1,2,9,3,1,5,2,1,3,4,7,4,2,2,5,6,5,2,6,8,6,3,7,3,1,3,1,3,8,9,3,1,9,9,3,9,7,4,6,9,3,4,4,6,2,0,7,4,8,8,0,0,4,0,6,6,8,9,9,3,3,2,7,9,1,0,3,3,7,3,7,8,8,4,0,9,1,4,6,8,3,6,2,8,1,1,3,3,7,5,6,2,5,4,0,0,4,5,1,9,8,6,6,4,5,3,9,3,6,3,2,4,8,9,2,0,2,7,4,9,2,8,1,0,6,0,1,0,1,9,9,7,5,8,6,5,1,2,8,3,6,5,2,0,3,2,0,4,7,5,5,6,8,3,9,9,2,6,6,1,5,6,7,1,3,3,8,3,4,6,2,7,2,6,9,3,9,5,6,1,0,1,8,1,4,7,4,9,8,8,4,4,1,0,9,1,3,2,8,7,6,4,7,3,1,5,4,0,9,0,5,3,8,3,6,4,4,5,2,0,1,3,4,1,2,6,8,4,8,4,7,3,7,3,1,2,5,8,2,8,8,9,4,8,5,1,3,0,6,2,5,4,3,7,3,7,6,4,0,0,3,5,6,6,0,2,8,2,8,2,7,7,9,7,3,3,2,2,2,2,8,9,3,5,9,8,0,2,1,9,5,3,5,0,8,9,7,3,9,4,2,8,4,4,9,1,6,7,5,7,4,5,8,6,0,4,3,7,9,3,2,8,6,4,3,4,0,6,4,8,4,7,2,6,5,4,0,2,6,0,8,7,0,0,8,8,9,4,4,4,6,1,3,4,5,9,8,6,2,4,1,5,0,0,0,6,9,8,6,9,5,1,1,7,2,8,3,8,1,8,8,1,8,2,0,7,2,0,9,8,2,8,0,8,5,8,4,1,5,5,9,5,9,7,8,8,5,2,7,6,3,0,3,6,9,6,4,7,8,1,0,1,4,1,7,2,3,0,9,9,6,6,4,0,5,9,2,9,2,6,1,5,8,7,4,6,2,0,0,9,2,8,6,5,7,2,6,5,2,7,6,6,1,4,0,8,7,9,7,2,3,4,2,1,8,6,5,3,0,1,8,4,5,2,4,7,6,2,8,8,8,8,0,5,8,9,6,8,3,3,9,3,3,1,5,2,7,3,2,8,6,1,0,4,5,5,5,5,8,5,8,4,7,7,8,0,2,6,1,6,8,2,6,4,8,7,0,4,7,7,9,3,4,1,4,4,8,2,0,1,7,8,5,4,6,9,8,1,1,7,6,1,7,0,2,5,8,9,1,7,2,7,1,4,4,6,3,9,5,1,4,2,1,8,1,7,9,2,8,0,1,7,1,8,7,6,1,1,0,4,4,8,1,1,5,9,3,0,0,8,7,4,6,7,2,5,4,8,0,1,0,5,2,2,6,3,2,0,3,3,8,1,2,9,5,3,1,8,0,5,5,3,7,1,5,6,0,3,8,8,8,6,1,4,8,0,2,2,3,9,2,4,2,6,0,6,3,3,1,7,2,1,3,6,5,8,3,9,5,6,0,8,5,1,5,9,9,6,8,7,7,5,6,8,4,5,5,7,2,0,1,2,5,7,1,3,9,1,4,6,6,7,6,3,8,6,9,4,2,3,0,4,2,4,3,8,6,4,7,6,7,7,1,5,8,6,7,7,8,7,4,6,2,3,1,6,5,5,5,8,4,0,1,6,9,6,5,3,5,2,3,3,3,0,3,0,5,3,8,3,1,0,7,8,7,5,7,6,3,8,1,3,8,6,0,5,1,1,4,1,5,1,4,1,8,3,1,7,9,5,6,5,2,8,0,4,7,0,7,2,6,8,6,6,9,4,2,6,6,0,3,7,2,6,7,3,3,2,4,8,5,7,2,9,7,9,2,9,2,1,2,0,4,5,7,8,1,2,6,9,8,8,1,1,0,1,6,2,1,7,2,3,9,4,8,2,3,2,7,1,0,8,6,0,0,7,3,9,6,4,1,2,7,9,8,4,2,8,4,0,1,4,4,8,8,1,1,2,0,2,8,7,4,9,3,3,1,9,8,9,7,2,4,6,1,0,3,7,5,5,0,5,0,3,7,0,6,2,3,8,8,7,3,7,8,9,1,0,2,4,9,5,8,4,9,5,4,3,6,1,8,5,1,7,2,6,0,6,3,6,0,7,5,4,0,5,3,6,6,5,8,8,9,5,7,9,4,3,1,0,1,2,6,0,2,2,5,2,4,4,1,8,1,3,6,4,9,4,6,7,8,9,6,6,0,3,8,1,0,2,2,7,9,9,4,7,6,1,9,2,9,9,1,2,4,3,5,8,3,0,9,6,3,1,0,9,3,9,1,2,6,9,2,2,6,6,5,5,0,3,5,5,5,5,7,8,3,1,6,9,6,5,1,0,2,6,4,7,7,3,9,7,6,5,1,6,6,6,4,7,5,9,7,8,0,5,1,5,5,0,2,8,6,1,5,4,9,4,9,2,9,4,8,7,0,2,8,4,6,5,8,5,5,6,0,6,1,6,4,4,2,9,4,2,0,8,7,7,0,2,2,8,2,2,7,7,8,1,3,2,1,0,5,3,0,8,7,1,1,4,2,8,1,7,0,5,7,7,7,6,7,3,9,6,6,2,6,9,4,9,5,2,4,8,4,7,3,1,4,2,6,8,9,6,0,4,6,3,0,1,0,9,1,8,6,2,2,6,6,4,4,5,8,4,1,2,9,8,7,9,0,1,8,1,0,1,5,3,6,7,8,2,9,2,4,5,0,1,7,3,3,5,4,9,4,5,6,4,4,6,9,3,6,5,9,8,1,9,5,5,7,2,4,9,1,3,4,9,3,3,1,2,7,8,8,7,2,4,6,6,1,5,3,8,2,4,9,3,0,6,0,1,2,8,7,5,0,8,6,0,7,8,8,4,4,6,5,1,3,3,9,3,8,4,6,9,6,4,2,9,4,7,5,1,4,8,0,3,5,1,0,4,8,3,9,9,9,0,1,4,7,1,2,7,1,6,9,5,6,5,7,5,9,3,3,5,3,7,1,6,1,3,8,1,7,0,6,1,3,1,0,1,7,5,5,3,2,4,1,3,4,6,2,3,8,5,9,7,6,4,7,1,7,3,7,7,0,5,9,0,7,3,8,1,8,2,7,4,8,4,0,0,2,3,2,7,1,4,2,4,7,3,4,6,4,5,6,9,3,6,3,3,0,8,9,4,0,9,7,5,6,4,1,7,4,0,6,8,3,5,3,8,3,2,7,6,0,7,9,5,0,9,9,7,4,0,7,2,6,1,9,9,0,1,9,2,1,9,6,7,3,3,4,6,2,4,7,6,2,8,7,7,8,5,9,1,6,4,8,4,8,3,4,3,6,9,1,4,8,7,9,3,2,4,2,6,1,4,6,7,5,3,2,1,1,0,9,9,2,7,8,6,8,6,3,4,5,2,4,0,0,4,6,4,5,5,8,7,0,2,6,1,8,5,0,2,9,0,6,9,5,0,2,8,3,1,2,8,9,8,4,6,0,5,1,3,7,5,8,1,2,6,2,0,2,0,6,2,6,8,6,0,9,5,4,8,2,0,5,6,1,8,9,8,2,5,7,2,3,9,4,6,3,9,2,9,2,2,3,4,6,7,1,7,2,8,1,4,2,0,9,5,1,0,0,0,1,6,3,8,4,1,6,8,4,3,4,3,0,1,2,0,8,9,3,3,5,3,7,7,3,8,7,3,4,2,2,2,4,5,2,3,2,1,7,4,7,3,4,3,2,5,9,3,3,3,8,3,7,8,5,8,3,8,6,2,2,0,5,5,2,4,3,7,6,6,8,6,4,2,1,6,6,1,0,0,6,0,7,8,5,6,2,9,2,5,2,1,8,7,3,2,8,0,5,4,9,4,1,2,0,5,1,2,0,4,6,3,2,3,2,4,0,8,6,7,8,3,3,6,7,3,2,4,7,2,2,1,3,9,8,6,0,8,0,1,8,8,4,0,5,5,8,5,7,3,3,8,8,4,7,5,7,8,9,6,8,3,3,4,2,7,5,5,5,1,2,7,3,7,2,1,3,3,9,2,3,6,4,0,4,2,5,7,7,8,3,6,6,0,3,3,7,4,7,5,7,8,6,9,9,5,4,6,7,5,5,7,3,5,2,1,1,3,0,6,3,2,7,0,3,1,9,4,7,4,5,9,2,7,8,0,9,8,4,2,6,5,8,0,7,3,0,0,9,7,6,7,3,8,0,6,7,6,4,8,5,3,4,1,5,1,9,8,6,2,4,4,0,3,4,8,2,8,5,2,7,9,2,1,5,9,9,2,2,6,2,3,4,3,6,8,8,0,6,8,2,0,1,3,4,5,4,7,8,8,5,9,9,1,5,6,7,9,2,3,3,9,1,0,2,3,3,1,8,2,0,4,5,3,4,3,8,4,9,8,0,4,6,0,6,3,5,5,6,9,4,6,6,3,8,1,8,4,1,1,2,6,3,6,7,1,6,1,6,5,8,6,4,9,1,4,8,5,8,1,5,6,4,3,1,7,3,2,0,6,2,5,0,1,5,1,6,0,0,2,0,8,6,0,3,2,9,6,8,2,2,9,1,2,5,2,7,6,3,4,9,3,9,8,4,2,5,7,4,9,7,4,0,7,1,8,6,4,8,2,9,8,4,7,8,2,9,5,2,5,8,2,6,3,3,9,4,4,5,1,4,8,7,9,3,5,8,3,2,8,9,4,6,4,0,2,0,8,9,7,6,1,8,0,7,4,2,3,3,5,9,9,5,2,8,0,9,9,8,5,1,3,4,3,0,8,6,9,3,7,8,5,1,8,9,7,1,8,1,9,4,4,6,7,1,8,4,2,2,1,1,5,2,4,4,7,4,5,5,7,8,8,2,6,0,0,2,7,2,5,1,6,8,4,4,5,3,4,8,1,5,5,9,0,0,0,5,4,4,8,8,9,6,4,6,4,5,9,3,8,3,0,2,4,0,0,0,7,9,2,4,7,8,6,3,0,3,2,6,7,2,5,9,2,9,6,3,4,9,3,1,3,1,1,7,7,1,9,7,5,3,8,2,2,0,3,7,8,9,9,5,6,8,9,6,0,5,7,1,9,3,9,7,5,7,6,3,0,6,1,5,3,7,6,5,7,2,3,6,0,2,1,1,7,8,7,2,7,5,5,1,0,3,4,1,8,0,2,1,5,2,1,9,4,1,7,7,1,2,0,4,4,5,3,8,7,2,1,9,5,2,3,1,8,9,5,2,5,4,0,8,4,7,0,3,7,5,5,7,8,6,3,8,7,4,1,0,6,4,3,2,3,7,7,4,6,6,6,0,1,7,1,1,0,3,6,1,8,5,6,9,9,7,2,3,1,1,1,3,9,3,3,8,8,3,1,0,4,4,9,1,6,3,6,1,1,8,4,1,9,1,1,7,6,4,5,7,1,4,2,1,6,3,8,6,3,7,4,5,9,3,3,9,1,9,0,9,2,2,2,2,3,4,3,7,5,4,0,7,2,4,1,2,3,0,0,0,7,1,7,2,5,3,2,0,9,6,6,9,3,2,0,7,0,2,8,3,8,7,8,4,0,7,6,7,6,4,3,9,6,7,5,8,7,2,8,5,0,4,2,0,4,2,7,3,8,5,4,1,3,3,3,7,9,1,7,9,7,6,5,0,2,2,3,1,2,2,9,1,8,9,7,5,3,5,3,6,1,2,9,5,7,9,4,3,4,5,6,7,7,9,3,0,9,6,3,1,5,1,8,2,1,9,3,2,4,8,1,4,1,6,5,5,5,8,5,3,8,2,6,4,5,8,8,1,5,8,6,1,8,6,8,7,3,5,2,2,3,6,8,9,4,9,2,8,7,4,7,9,6,8,5,3,4,8,0,2,3,2,3,5,8,8,9,0,3,7,1,1,7,3,0,8,0,7,0,6,2,1,3,7,0,3,2,1,6,8,6,0,7,8,2,2,7,1,1,8,8,1,4,9,7,0,9,2,3,3,8,3,8,1,7,1,2,6,3,5,1,6,1,9,9,9,0,8,1,9,0,7,9,4,7,9,5,3,4,5,8,8,6,1,0,7,2,0,4,3,0,0,0,5,7,2,0,4,7,9,9,4,6,3,7,4,7,2,9,1,9,5,8,1,8,7,4,8,8,8,2,8,4,3,7,9,9,1,1,2,5,4,2,1,1,9,1,0,2,7,1,6,5,0,5,8,6,8,2,4,9,4,4,5,6,4,3,9,0,4,3,7,0,2,6,5,7,4,2,6,7,6,7,6,3,9,3,0,7,8,9,5,1,7,0,4,3,5,8,5,3,4,4,3,4,3,6,6,5,1,9,3,0,9,9,5,9,9,4,0,1,6,3,2,9,9,2,4,3,6,1,7,5,5,2,3,3,2,6,4,8,2,1,3,4,1,4,7,4,1,7,7,9,3,6,0,2,6,1,5,9,2,7,2,8,9,2,2,6,3,5,6,0,2,5,3,5,3,2,2,9,0,3,4,5,4,3,9,7,6,9,4,0,0,6,9,2,8,3,6,7,8,7,0,5,9,1,1,8,9,0,0,6,9,7,5,6,3,5,3,8,6,3,7,2,5,9,1,2,8,0,4,7,5,0,3,1,7,1,3,6,4,2,8,0,1,2,4,2,1,8,4,0,0,0,7,4,7,8,8,6,6,5,5,3,5,8,5,0,1,8,1,6,1,9,5,4,7,1,9,1,7,3,2,0,3,9,0,5,6,8,1,6,4,2,6,9,5,9,5,1,7,0,2,1,0,5,7,0,8,8,2,7,5,6,4,6,4,4,6,4,3,5,6,3,2,4,7,1,6,0,2,8,9,7,5,5,4,5,3,0,3,4,0,3,4,8,3,4,5,6,3,5,5,8,1,9,5,3,3,9,6,1,0,9,7,6,3,4,3,7,5,4,7,8,9,7,3,6,7,3,7,6,6,4,6,3,0,1,0,4,6,6,5,3,8,3,9,5,6,8,4,5,1,4,2,6,6,1,4,6,8,0,3,9,0,6,3,5,2,7,1,6,5,3,5,8,0,2,0,7,8,7,8,2,1,2,8,0,0,9,6,6,0,0,7,8,8,7,9,7,5,0,8,3,8,6,6,6,1,5,0,4,2,4,1,4,7,1,2,9,6,5,3,5,8,0,4,0,4,3,5,7,8,7,3,8,7,4,9,6,1,4,0,7,7,6,9,7,9,4,4,1,5,1,7,0,0,6,3,4,9,9,4,4,8,9,0,9,2,4,6,4,9,7,6,6,8,2,2,9,5,3,4,9,7,4,5,3,8,3,7,7,8,9,1,9,9,9,1,0,9,8,4,1,4,7,8,7,5,7,5,9,7,9,4,8,0,7,1,7,9,4,3,2,2,9,8,9,9,2,9,4,1,2,4,3,0,2,4,6,7,5,8,4,1,6,2,4,1,7,0,8,0,1,0,6,4,1,1,5,6,0,9,0,2,2,3,4,2,5,2,1,8,1,8,7,8,1,0,1,7,8,2,3,3,4,5,8,3,4,1,6,0,6,4,4,9,3,5,4,2,9,1,0,4,9,8,2,0,2,7,9,9,2,6,3,3,7,3,0,5,2,6,9,0,3,9,1,1,4,6,6,9,9,8,4,0,7,8,2,1,3,3,8,7,1,9,7,8,4,1,2,1,0,3,1,7,0,0,1,2,9,1,6,0,1,2,0,7,5,8,7,4,5,3,5,0,1,2,6,1,9,9,2,2,0,8,9,9,3,6,6,8,8,3,1,3,5,8,1,6,6,0,2,0,2,3,2,1,0,7,9,2,3,6,3,9,6,7,3,2,3,3,5,0,6,7,5,8,4,1,2,2,4,7,4,1,4,8,3,5,5,0,7,1,7,0,3,5,9,0,5,3,5,6,3,3,4,4,5,2,9,5,8,8,4,1,2,2,4,7,8,1,7,1,2,6,6,7,7,6,4,2,6,0,8,8,0,9,9,0,8,9,9,9,4,2,6,9,0,9,9,0,5,7,6,4,1,9,4,3,2,6,9,2,6,2,2,9,9,0,9,0,7,2,9,5,5,9,2,9,7,5,3,3,9,5,7,6,2,0,2,3,2,6,0,0,5,0,5,3,9,9,6,3,3,0,3,8,6,9,2,3,5,3,1,8,9,6,7,5,8,0,2,6,7,8,8,8,8,3,5,3,2,5,3,7,6,5,5,2,7,9,4,6,6,7,5,2,6,3,1,9,4,3,9,4,1,0,5,9,2,4,9,9,1,4,5,9,6,3,2,4,2,6,2,0,2,2,5,3,7,2,9,9,2,3,3,3,7,6,0,0,2,7,7,2,3,4,5,6,4,6,5,8,8,6,9,9,0,6,1,4,0,4,6,4,5,8,7,8,7,3,2,5,6,9,5,2,1,8,0,8,2,0,3,6,7,9,9,8,2,2,2,5,2,4,5,7,1,6,3,4,8,7,8,6,3,4,5,2,0,1,4,7,4,0,7,8,5,7,2,1,3,0,5,6,8,3,3,1,5,1,8,4,0,9,9,0,5,6,3,8,9,2,1,3,8,0,6,5,8,3,4,5,5,8,0,6,4,8,2,4,1,0,2,5,1,8,9,3,9,3,7,1,8,0,5,7,3,6,6,1,4,4,6,3,8,5,0,7,0,6,7,6,5,5,2,5,1,8,6,0,8,7,4,6,5,8,8,3,7,0,0,4,8,9,2,4,1,7,6,4,0,0,6,6,5,4,0,9,2,7,0,1,2,9,7,6,8,8,8,3,2,6,4,7,8,2,4,7,3,5,3,4,6,2,0,4,2,6,4,8,5,5,9,3,0,0,6,8,9,2,0,4,6,1,2,0,3,7,9,6,0,2,7,6,4,9,4,4,7,0,7,4,5,6,7,7,1,0,9,9,9,6,2,5,2,9,7,1,4,8,9,2,7,9,6,0,0,6,7,4,4,5,7,0,8,6,3,6,5,6,8,9,7,4,0,7,4,7,3,8,5,6,8,9,6,0,5,6,8,4,1,0,4,4,4,6,4,5,4,9,1,8,3,1,3,8,4,1,8,9,0,1,0,2,0,1,2,7,0,9,9,6,1,1,0,2,4,2,3,1,0,8,9,7,8,3,9,4,0,4,0,4,0,0,1,5,6,1,1,5,3,0,1,7,1,6,0,7,7,6,5,1,8,7,3,1,2,8,9,6,5,0,8,7,8,8,2,9,1,9,3,7,6,0,6,2,1,2,7,6,9,8,2,3,0,1,7,9,4,7,6,5,2,8,4,8,9,9,5,4,8,8,6,4,6,4,9,7,0,1,1,1,4,0,0,3,9,6,8,3,3,0,3,0,8,6,4,9,9,0,1,9,6,2,7,6,5,4,9,3,7,2,0,8,9,5,3,9,7,2,2,0,0,3,9,4,4,7,5,8,4,7,4,1,7,1,4,0,7,0,9,2,9,9,8,0,7,6,4,8,8,5,9,8,6,5,9,9,1,9,7,5,4,3,5,1,9,6,3,0,3,4,4,5,5,3,3,0,3,8,8,4,7,3,8,3,0,2,9,7,3,3,6,6,0,4,1,5,5,9,1,3,4,7,2,7,9,6,6,9,4,1,9,1,7,2,0,3,8,6,2,6,9,4,2,7,0,4,5,0,7,9,5,7,7,7,7,0,2,6,7,1,3,9,0,5,6,0,2,9,5,7,0,4,5,6,1,8,7,3,0,8,2,6,3,3,1,8,8,6,3,7,3,8,0,7,7,3,2,1,9,4,3,9,7,3,9,6,3,1,2,8,1,4,6,7,3,4,6,2,0,6,0,3,9,9,1,4,6,2,3,6,4,9,5,1,4,6,6,7,7,5,2,1,2,4,3,8,6,9,8,3,0,1,6,8,3,6,4,8,7,0,7,0,4,8,3,8,0,5,7,5,5,6,5,4,5,8,2,3,5,7,8,2,7,8,3,2,0,1,4,6,8,0,5,2,5,1,8,3,1,8,2,6,5,2,6,3,2,6,4,0,0,1,1,2,5,6,6,7,6,4,2,2,4,5,8,2,1,6,8,2,5,6,3,4,4,6,7,3,2,1,0,1,9,3,6,5,0,5,6,0,6,6,1,1,6,4,3,3,6,6,4,7,8,6,6,5,5,1,9,1,7,5,9,8,7,4,1,3,2,9,2,7,3,4,7,2,9,8,4,5,5,4,0,0,5,4,1,9,2,6,9,1,9,9,2,0,8,5,2,8,3,0,7,0,1,9,1,8,0,7,9,2,6,1,5,0,6,1,7,1,9,4,3,9,9,5,3,5,3,3,9,6,0,2,5,9,3,2,1,2,8,7,2,2,0,0,7,7,6,8,8,8,9,7,7,4,2,9,0,0,4,3,0,5,3,3,7,5,4,4,0,0,2,4,5,8,1,1,0,3,6,8,9,1,3,5,3,1,6,2,6,7,4,7,7,3,6,3,4,3,7,3,2,1,6,1,9,6,7,1,4,5,7,4,1,5,9,9,4,3,4,2,2,6,5,8,5,0,3,4,4,8,4,3,3,7,2,7,6,3,3,7,9,6,6,6,6,3,3,3,6,3,4,0,5,0,9,0,9,5,1,4,0,9,3,5,3,0,5,8,1,5,7,5,8,9,5,6,3,7,1,9,7,1,0,5,5,3,8,3,9,8,4,5,4,7,3,0,8,9,1,3,8,4,8,1,0,2,9,5,1,3,7,9,2,9,4,6,1,7,5,7,1,3,0,6,1,6,4,3,2,5,7,7,2,9,7,3,4,1,5,3,2,3,8,6,8,3,9,2,6,1,0,8,1,1,4,7,3,1,3,3,3,0,6,5,3,4,7,0,6,5,7,0,4,9,8,5,7,8,3,0,3,2,6,5,0,1,2,9,1,5,0,9,2,4,8,0,2,8,0,2,0,4,4,8,7,9,6,0,2,5,0,3,0,8,4,4,4,2,7,4,4,2,7,3,0,1,7,8,5,4,5,3,6,1,2,8,5,7,6,9,3,0,2,4,0,6,2,3,4,7,7,0,5,8,5,8,0,1,5,0,2,7,5,1,4,5,6,6,4,8,3,5,2,5,3,8,9,4,1,5,0,3,2,3,2,7,7,3,0,3,6,8,4,4,4,4,4,5,2,9,6,3,4,2,2,0,6,1,0,4,9,4,6,7,6,5,5,9,0,4,4,6,9,8,2,7,6,1,6,1,9,1,3,3,7,2,0,2,1,8,6,4,1,4,5,5,0,4,2,7,0,1,3,7,0,2,7,5,9,2,6,5,9,0,0,0,3,5,8,6,9,0,1,8,0,4,7,8,5,1,7,1,6,1,5,3,5,5,4,5,2,2,9,9,7,3,0,9,3,7,6,9,1,0,2,8,7,8,4,0,2,4,3,2,0,1,6,8,2,0,5,4,0,6,3,2,8,6,8,8,1,4,2,6,1,3,8,7,9,8,8,3,2,0,4,4,2,1,8,8,7,4,4,7,8,9,3,3,0,3,6,5,1,4,6,7,9,4,5,5,3,2,9,0,6,2,1,8,5,3,0,2,2,1,3,0,6,5,8,4,6,3,8,0,4,1,6,7,2,3,7,8,9,3,6,5,1,7,3,9,0,9,1,3,3,7,9,8,9,6,5,3,4,0,5,0,9,0,4,4,8,7,8,6,8,3,1,6,5,0,5,4,7,8,5,6,7,1,0,9,0,3,0,1,8,4,0,5,5,2,4,6,6,3,1,4,9,4,3,8,3,6,4,7,4,6,8,9,9,2,8,4,3,1,0,0,9,1,3,5,2,7,3,7,9,6,3,2,9,4,4,0,2,4,1,4,0,2,5,1,3,3,7,0,9,4,5,0,5,5,1,7,7,7,7,0,9,5,6,4,6,7,8,5,0,2,2,3,2,5,4,8,1,7,5,7,5,3,2,4,3,7,7,3,8,2,4,4,2,7,4,1,7,1,1,6,7,2,7,6,4,1,8,3,0,6,3,1,5,0,1,5,5,6,5,3,5,1,0,8,7,3,0,0,7,7,9,2,9,0,7,1,9,3,6,6,0,2,9,5,6,7,5,8,1,9,5,7,0,3,1,0,6,5,7,7,8,4,3,6,6,3,8,7,7,0,7,3,7,1,2,2,4,4,0,2,4,2,6,8,3,3,9,0,4,5,2,3,5,8,1,2,3,7,2,2,8,7,9,7,2,2,1,8,8,5,6,4,4,7,7,8,9,2,6,9,6,0,5,9,2,0,0,7,2,1,5,0,1,5,4,7,6,6,1,8,2,4,0,3,6,1,4,2,1,8,5,5,3,9,0,8,7,3,8,1,0,1,5,0,1,2,5,4,8,3,4,8,6,3,4,5,3,1,3,6,0,2,5,8,1,0,1,4,3,1,8,6,3,5,6,9,1,5,8,3,4,0,8,2,3,9,6,9,3,0,3,5,3,9,3,9,7,7,7,1,8,0,0,5,6,2,2,6,6,3,3,4,2,7,0,4,8,3,1,3,2,8,5,1,3,7,5,8,3,7,5,5,8,8,3,8,8,7,6,8,3,8,1,2,3,6,4,6,1,1,3,1,9,3,0,6,1,9,1,7,2,6,3,0,0,5,6,1,8,6,1,7,9,0,4,5,7,5,0,2,9,4,4,8,2,3,0,3,2,3,3,4,9,8,6,5,0,9,8,7,2,9,7,7,8,4,0,1,1,8,4,7,9,0,2,2,9,6,9,6,7,3,1,1,3,7,5,5,3,8,6,9,6,1,9,2,4,7,0,5,0,3,7,4,0,0,4,2,2,9,0,1,3,0,7,7,2,0,5,3,1,5,4,2,4,0,3,6,8,1,6,0,8,8,3,0,5,8,8,3,9,7,7,9,9,5,9,6,7,6,6,1,8,9,6,7,8,5,4,4,7,5,7,2,9,4,6,2,3,0,4,8,7,8,8,8,9,2,9,9,0,7,6,1,4,9,7,2,4,4,2,7,8,4,4,4,2,6,9,9,3,1,8,6,5,1,2,4,9,2,6,4,8,2,1,5,9,3,5,1,8,3,2,0,8,8,5,2,7,7,8,0,0,2,9,1,2,6,5,1,6,5,9,5,5,7,0,8,5,9,4,5,3,2,5,2,5,5,4,9,5,6,8,1,9,7,7,8,4,6,6,2,6,4,9,8,6,2,8,3,0,4,6,4,4,2,0,6,7,5,5,5,5,8,1,3,9,6,5,9,3,2,0,8,9,8,0,2,5,6,0,6,5,8,5,7,4,7,6,0,5,8,4,2,4,4,0,1,9,3,5,5,5,1,5,7,5,6,1,4,6,8,1,4,4,6,1,9,7,9,5,2,6,9,3,6,5,7,5,7,7,1,2,9,2,5,1,2,4,7,0,8,8,3,8,7,6,8,9,3,7,5,6,5,9,8,4,5,1,6,8,4,8,5,2,9,5,0,8,9,7,2,9,3,8,6,3,5,3,8,6,8,1,2,8,3,6,0,4,7,8,6,7,1,9,5,4,5,0,6,9,0,1,0,2,9,6,5,7,0,2,7,6,2,4,6,3,8,9,9,3,4,7,2,2,5,0,9,5,1,9,2,0,9,3,5,4,5,1,5,4,5,5,6,2,3,1,5,2,6,0,1,3,6,2,6,8,2,3,3,1,0,0,3,4,6,2,4,5,3,5,2,7,1,1,4,4,5,6,3,0,7,1,3,7,0,2,1,1,1,3,9,9,9,6,2,9,8,6,1,4,6,7,0,9,0,0,4,3,3,4,0,4,4,1,8,7,1,0,3,4,2,2,9,2,0,8,3,3,5,3,9,6,6,5,7,3,1,9,6,6,8,4,9,9,1,1,2,7,9,1,5,7,1,6,3,9,1,0,6,5,3,6,3,2,5,5,7,5,7,4,4,8,1,5,7,7,7,7,2,5,7,1,1,9,1,3,6,4,6,6,8,4,6,3,1,8,2,6,1,6,2,3,7,2,9,0,5,4,0,1,0,5,6,7,7,3,6,8,9,6,1,3,8,1,1,1,1,6,3,3,7,5,5,5,1,8,2,8,5,9,4,7,9,1,1,2,6,5,1,3,3,0,3,2,6,5,9,1,2,9,3,5,3,4,6,2,5,1,2,3,6,1,1,2,7,2,2,0,0,0,2,5,6,1,8,3,0,9,8,2,0,6,2,4,7,0,4,2,9,9,2,4,6,6,1,2,6,5,3,8,7,4,7,3,6,9,2,2,0,3,7,8,6,0,7,4,8,0,9,1,2,0,3,6,7,4,8,1,0,7,8,4,9,0,1,1,9,8,4,7,6,9,0,1,6,1,2,0,7,3,1,1,0,2,6,3,0,0,9,0,5,4,4,4,7,7,1,9,1,5,1,6,0,1,8,5,8,5,1,7,7,5,5,9,1,3,6,3,8,0,9,3,1,8,5,7,1,7,9,1,2,1,5,4,4,6,1,1,2,0,7,6,0,7,1,3,5,1,7,1,0,2,1,4,3,6,2,5,4,2,5,3,9,1,6,2,5,3,5,6,4,5,8,0,9,8,3,5,7,2,4,9,6,5,7,7,2,6,1,4,3,0,2,9,0,1,3,6,9,1,5,9,7,0,9,3,3,4,9,2,3,4,8,7,9,3,0,1,2,6,2,9,8,0,8,6,8,8,7,7,7,5,4,6,5,6,5,0,7,4,7,2,0,6,8,4,1,2,5,0,0,7,5,9,3,9,9,8,6,2,5,3,6,0,3,0,3,8,6,2,4,6,3,4,7,2,7,9,2,2,3,8,0,6,5,4,0,9,3,3,4,9,9,0,4,1,3,1,6,7,7,3,3,7,7,6,0,7,8,8,2,8,6,2,7,4,0,5,0,6,3,4,0,8,8,0,6,0,4,8,5,3,4,0,9,6,6,1,5,4,7,8,6,1,5,3,3,7,9,8,8,3,7,3,8,4,3,4,8,4,4,1,0,0,4,0,6,3,2,8,1,5,1,1,8,4,3,0,7,2,0,7,1,0,9,1,5,2,9,9,8,9,4,9,7,3,2,6,2,6,1,2,7,9,3,5,0,6,1,3,9,7,3,1,6,3,6,3,5,1,1,5,3,3,9,1,4,2,4,9,8,9,2,6,1,4,7,4,2,0,3,6,6,0,0,6,0,1,3,0,8,2,3,8,6,0,6,9,0,9,5,0,2,0,6,5,7,1,3,1,8,7,7,1,6,5,9,5,8,3,6,6,0,3,2,2,7,7,7,4,9,1,3,5,8,2,7,3,1,4,3,3,0,7,2,5,2,9,8,8,3,9,4,0,2,3,1,1,5,3,5,5,7,8,1,0,0,5,2,9,6,7,2,0,0,7,8,7,3,6,8,3,9,3,0,4,3,4,5,6,7,8,8,6,1,6,1,9,7,3,2,8,4,8,3,3,7,1,8,7,9,4,4,1,5,6,9,7,6,7,6,4,7,3,4,0,1,3,7,4,8,8,8,5,8,0,1,5,9,2,1,5,8,5,4,4,8,9,7,0,6,5,9,8,0,0,1,7,8,6,0,4,7,4,4,1,1,3,9,8,5,6,0,1,5,0,1,8,4,0,9,8,9,8,3,7,2,8,9,7,3,4,7,2,3,8,6,8,9,7,0,2,8,5,1,4,0,9,7,5,3,8,4,4,1,7,0,0,1,7,4,3,7,1,1,5,5,6,1,3,5,5,4,2,3,0,1,7,6,3,3,6,3,6,5,1,4,1,8,9,5,5,3,4,0,1,8,3,7,3,7,4,5,0,5,3,0,7,5,4,5,9,4,0,0,7,7,6,1,1,5,5,8,3,9,5,8,5,2,5,6,4,9,9,3,9,3,6,1,2,9,1,9,7,8,4,6,5,8,0,5,7,8,2,7,2,3,6,7,9,9,6,9,6,3,5,1,4,7,1,4,9,4,0,9,5,4,4,0,3,3,6,0,4,6,0,5,3,8,1,3,2,0,3,0,6,7,4,8,9,9,0,4,6,2,7,9,5,2,3,1,3,2,1,6,8,8,7,3,6,9,3,4,0,0,1,3,9,5,3,9,3,6,0,3,9,9,4,0,1,1,6,6,8,2,0,0,1,8,3,3,0,0,4,1,7,3,0,2,8,6,2,5,3,5,9,4,7,2,5,4,6,2,1,7,9,6,0,1,2,5,8,9,6,1,9,2,7,2,1,0,8,1,9,5,3,6,0,7,5,8,7,1,8,6,0,3,5,4,6,2,6,0,3,2,5,6,8,1,9,3,1,6,0,5,3,3,6,4,2,1,4,1,2,4,8,8,2,8,5,3,5,1,0,8,6,2,5,7,8,0,9,3,8,3,3,1,9,3,7,2,7,5,3,1,4,6,8,4,5,8,3,3,4,5,7,3,7,5,2,8,3,3,3,0,1,0,7,2,1,7,1,2,8,8,2,8,4,3,0,0,1,8,5,6,4,0,9,3,6,4,2,2,1,4,3,3,2,0,5,6,7,2,4,6,0,0,7,7,0,6,4,3,5,0,8,1,0,5,9,7,6,5,1,5,1,0,1,6,5,7,8,5,1,3,1,1,2,9,5,0,7,0,0,9,8,5,5,4,2,6,3,5,7,3,1,2,7,7,8,2,4,9,9,6,5,9,5,4,5,4,5,5,3,9,4,7,5,4,6,6,0,3,9,2,4,9,5,1,2,6,4,7,9,4,6,9,0,2,6,2,4,5,9,9,5,2,6,2,1,3,6,9,8,8,8,5,9,4,3,7,0,8,2,5,1,9,4,4,3,1,8,9,4,7,1,6,6,6,0,6,5,1,6,1,9,2,8,4,6,5,5,8,6,8,6,0,5,9,7,9,8,1,7,5,3,4,8,2,2,1,4,0,5,3,4,9,2,7,1,1,0,3,0,4,4,7,3,0,6,2,9,9,6,0,5,6,3,6,2,6,0,6,5,0,4,1,0,9,4,9,5,8,2,1,7,1,0,5,7,7,2,1,6,3,7,6,3,2,1,9,7,4,5,5,7,4,5,4,7,8,0,2,2,8,0,7,1,8,7,0,9,3,6,8,3,2,1,2,9,9,8,6,0,4,5,6,2,9,6,6,0,5,1,2,6,7,1,9,0,5,6,8,7,1,0,0,2,7,1,9,6,1,9,7,2,3,2,5,3,9,6,6,9,3,6,1,8,7,2,4,6,6,5,7,1,8,2,8,9,8,3,0,9,3,5,4,4,7,9,4,7,3,6,8,9,3,5,7,1,8,0,9,2,4,3,8,3,3,0,1,8,4,9,4,6,6,6,6,7,1,3,5,6,8,3,4,1,5,3,8,6,1,0,5,4,3,6,6,0,5,5,3,7,5,4,8,7,1,6,9,9,5,6,0,0,8,1,0,6,7,3,5,9,0,9,9,0,1,8,8,7,6,9,2,9,4,6,2,3,6,1,3,5,0,1,0,8,4,4,0,1,3,8,9,7,7,5,3,4,1,7,1,3,4,8,4,8,6,0,4,1,3,3,8,2,5,6,5,9,3,4,4,0,0,6,4,9,8,1,9,6,6,9,9,1,4,4,8,0,3,7,9,0,2,1,9,3,8,8,3,2,5,2,9,5,1,8,0,1,0,1,4,5,9,8,3,2,5,7,1,3,8,8,6,6,3,6,5,5,8,4,9,6,3,2,8,7,6,3,2,7,5,7,2,2,5,9,1,9,4,5,3,0,1,6,8,4,4,8,4,5,4,2,3,7,2,5,0,6,5,9,0,7,9,0,8,0,4,3,6,9,3,4,0,7,1,7,2,3,4,3,3,6,8,7,3,7,9,4,6,6,9,8,8,2,4,5,6,4,0,8,3,2,6,2,0,0,0,2,6,7,9,6,1,6,9,4,2,0,9,3,9,9,0,2,7,3,9,6,3,1,4,2,7,0,3,4,7,7,6,4,7,4,2,6,5,7,0,8,0,9,5,3,1,8,7,1,1,5,2,9,7,0,4,2,2,2,2,6,7,6,9,8,9,6,9,7,2,0,2,1,6,4,2,7,9,2,0,5,0,3,0,8,5,6,8,0,6,9,1,0,5,7,9,6,1,1,9,3,9,9,8,7,2,2,2,7,2,0,0,1,8,4,6,0,1,2,0,2,9,9,9,4,1,4,5,6,4,5,2,2,6,8,1,3,0,5,0,8,7,2,0,0,4,4,4,6,3,8,0,4,7,2,0,0,5,7,5,6,1,0,7,0,2,1,2,1,3,9,0,1,1,4,6,5,8,2,1,5,4,0,1,3,5,6,7,9,8,3,4,4,0,6,1,7,8,9,8,5,0,3,2,5,7,7,7,0,9,9,7,3,4,3,8,4,1,1,3,7,7,2,7,0,3,0,8,3,9,4,7,9,0,0,0,9,7,7,0,2,9,1,2,4,0,7,1,9,3,2,8,5,5,4,3,2,1,2,6,3,8,5,0,2,6,1,3,8,0,7,0,4,2,7,0,4,4,8,7,9,8,2,9,4,1,8,1,6,2,1,7,4,2,5,0,2,1,0,2,6,4,9,0,8,4,1,3,6,2,0,8,0,6,0,6,9,5,0,2,2,6,5,2,5,4,2,4,8,6,9,4,3,2,3,5,9,0,4,2,1,3,7,9,8,4,5,2,4,8,4,0,1,2,3,1,8,7,2,2,3,5,3,6,9,4,2,4,6,8,0,8,0,6,0,3,9,9,7,2,0,2,2,0,5,9,6,0,3,7,5,6,7,8,7,8,5,3,4,7,1,2,9,1,2,9,8,4,4,7,9,3,6,7,9,7,3,6,4,5,0,8,9,6,6,4,7,9,3,7,0,4,6,9,7,1,1,8,3,4,4,8,2,2,8,4,7,8,7,6,3,7,3,1,8,7,9,7,6,5,9,8,0,3,9,2,2,9,9,0,8,3,6,8,9,2,7,0,0,7,7,0,4,6,5,3,4,0,9,7,0,5,5,2,6,9,9,0,5,1,5,3,1,0,0,3,1,2,4,7,9,0,8,3,8,4,4,2,2,0,7,8,1,0,6,2,4,5,8,4,9,3,8,2,2,9,5,6,6,5,1,1,4,8,8,3,6,6,0,4,9,3,4,9,4,8,2,6,1,6,6,2,8,0,5,6,5,5,6,5,4,3,0,1,8,0,2,5,5,0,6,3,6,4,8,9,4,2,9,7,8,3,5,4,8,7,4,8,1,3,8,6,5,6,1,4,0,1,5,6,2,0,6,3,1,4,4,3,3,5,9,9,9,5,8,9,9,5,2,8,8,0,7,2,6,5,9,0,9,6,3,6,4,7,5,6,3,3,4,8,5,2,3,4,4,1,5,3,7,7,3,0,3,5,8,1,1,5,1,8,0,8,9,6,2,5,1,1,7,7,2,9,6,6,5,3,0,2,2,3,6,3,4,3,9,6,3,3,4,0,8,4,7,9,9,1,8,5,7,8,5,0,1,1,0,1,0,5,2,7,1,3,0,0,2,8,9,8,2,7,5,2,0,2,8,1,1,3,8,0,5,9,7,1,4,3,9,2,9,8,1,6,2,2,3,0,2,2,9,7,4,1,8,6,6,7,3,3,0,2,6,2,4,6,3,5,8,9,6,2,7,6,4,7,8,9,8,3,1,8,0,2,2,6,9,4,8,9,8,5,5,1,5,8,4,1,8,9,4,4,9,7,1,9,8,8,3,6,3,5,6,5,8,1,3,9,5,7,7,7,8,0,8,9,8,1,4,8,6,6,9,3,4,4,5,6,7,2,9,8,9,7,8,9,2,2,2,2,2,2,2,7,6,8,5,8,9,6,3,2,2,6,3,4,4,8,7,9,7,1,6,6,8,7,9,7,1,8,3,4,0,2,3,4,1,0,1,4,6,9,2,4,7,6,1,3,8,9,8,2,6,1,0,4,5,6,0,3,2,9,1,9,9,5,5,1,6,7,0,4,1,5,4,0,6,7,2,3,9,8,9,7,1,0,5,2,3,0,9,2,4,5,7,6,1,5,4,4,4,0,9,2,3,1,8,2,9,3,1,2,1,3,9,8,0,3,9,8,4,8,9,6,9,6,0,9,5,3,2,8,6,2,7,1,8,9,7,9,6,5,9,1,6,3,5,7,8,3,1,8,6,8,3,9,4,0,8,9,6,2,9,2,6,7,2,2,7,3,1,3,3,1,3,3,3,0,0,9,4,2,4,6,1,0,4,7,7,2,9,6,3,5,8,1,5,6,1,5,2,3,2,0,0,9,6,4,5,4,2,1,6,7,1,5,1,0,0,2,6,7,5,3,0,8,3,4,5,4,2,4,8,6,0,3,3,9,9,3,3,4,7,8,9,9,8,9,4,5,2,4,5,9,8,8,1,4,9,0,6,6,6,5,6,7,2,7,3,4,9,5,6,1,5,2,2,6,9,3,6,1,2,0,5,2,5,9,5,5,0,6,9,9,5,6,5,8,7,9,3,6,8,8,1,6,6,0,6,8,9,5,5,5,9,4,5,4,4,1,4,7,1,1,1,8,9,5,3,2,3,3,8,4,3,3,0,5,9,9,7,7,7,5,3,1,9,9,3,7,5,5,2,5,8,5,6,7,7,8,7,1,5,7,6,8,0,3,8,4,0,7,9,5,1,8,5,7,6,7,1,4,8,6,8,5,1,2,0,3,4,8,1,7,8,0,4,4,1,9,9,7,9,5,9,7,0,6,8,2,8,6,8,3,0,2,3,3,7,4,4,0,8,8,7,0,9,3,7,1,2,8,5,9,7,1,8,8,8,2,2,7,8,6,6,7,4,1,8,3,1,1,0,3,4,4,0,0,8,2,6,8,1,5,6,1,2,1,4,0,8,7,0,6,7,7,9,3,0,6,5,5,6,5,3,9,1,4,0,4,0,6,3,9,8,1,3,5,5,7,5,3,3,1,9,2,7,0,2,6,4,2,7,4,1,0,9,7,8,2,1,7,3,0,0,2,6,1,1,7,4,1,7,4,6,9,6,4,0,6,8,5,4,9,0,3,2,1,3,9,7,3,1,3,0,7,9,3,2,3,7,4,3,7,1,2,7,8,5,3,6,3,1,3,2,0,7,0,7,4,1,1,1,1,8,2,3,0,8,4,8,3,2,2,3,6,3,2,2,1,6,1,3,6,4,5,0,3,8,6,4,3,7,9,8,4,3,2,4,7,3,8,8,9,3,3,4,1,2,9,4,8,9,1,1,3,0,9,9,6,9,7,4,3,6,3,6,4,2,2,0,9,8,3,6,9,6,1,7,2,3,1,0,1,1,8,2,9,5,1,9,1,9,7,1,4,2,5,2,6,7,7,3,4,6,3,9,4,0,3,3,9,2,6,5,9,2,8,3,2,2,0,2,6,3,3,0,2,8,1,4,9,7,6,9,3,4,7,6,5,7,1,2,5,1,6,5,7,5,2,5,1,3,5,4,4,0,1,2,9,5,6,6,8,5,1,3,4,2,0,7,5,1,4,1,7,9,1,4,8,3,7,4,9,6,1,6,0,8,6,3,3,5,6,1,0,5,8,8,4,0,6,9,9,2,9,9,9,8,0,6,4,2,5,5,8,1,5,6,7,1,1,0,1,0,8,9,8,3,6,8,0,6,4,2,7,9,3,1,1,9,6,3,9,7,6,2,3,6,1,7,8,1,4,3,2,1,0,9,2,7,3,0,3,3,0,2,9,0,5,2,6,4,0,9,1,4,9,2,4,8,0,2,4,5,4,3,8,9,8,9,0,8,1,5,2,5,9,8,4,7,9,3,2,3,6,3,5,0,4,0,3,5,5,8,2,4,4,7,1,2,6,4,1,2,4,8,9,6,2,5,4,8,8,0,4,2,0,1,1,2,5,2,9,6,1,2,4,9,8,2,3,5,6,9,2,6,7,4,7,3,7,8,4,1,0,4,7,1,6,5,1,3,0,9,3,0,0,9,6,7,5,6,4,3,7,8,1,9,3,3,7,4,8,1,0,1,9,3,6,5,5,6,7,3,7,7,0,5,2,9,2,3,0,7,2,6,1,6,3,9,0,3,7,1,2,8,9,7,3,9,7,7,3,6,2,1,4,1,5,4,6,8,5,0,6,7,6,4,5,6,5,2,5,6,1,5,5,2,0,2,0,5,4,7,4,3,7,1,9,5,6,9,6,5,8,4,5,0,9,5,5,1,7,7,2,7,8,1,5,1,2,7,2,2,3,2,5,2,3,2,3,4,4,7,6,4,5,0,3,5,9,4,2,3,5,6,4,3,1,7,7,4,2,3,2,7,5,7,9,2,9,8,0,5,4,9,7,1,8,2,1,9,5,0,0,8,4,4,2,5,6,6,1,0,1,8,6,3,5,3,2,9,7,9,6,8,4,1,4,1,7,7,8,0,4,2,4,3,6,7,7,3,7,5,2,6,8,6,8,5,9,7,9,8,0,8,3,5,4,0,4,4,9,1,3,4,6,7,4,8,1,2,3,0,2,6,9,7,3,9,5,5,4,5,2,4,4,9,4,4,3,5,8,1,6,4,0,8,0,8,1,4,7,2,1,4,7,6,8,3,7,4,1,0,2,1,8,6,0,3,9,0,8,3,2,3,1,0,8,2,4,6,6,8,9,7,2,3,4,4,3,0,4,8,3,3,9,6,3,7,9,5,6,2,5,8,9,5,0,0,2,2,0,5,2,4,7,2,3,1,5,7,1,8,9,4,6,4,6,7,8,4,5,9,3,6,3,7,9,9,4,2,7,6,7,0,9,8,0,6,9,1,2,5,3,4,5,4,4,1,1,7,5,0,1,0,4,6,8,0,5,8,8,7,9,4,1,4,0,0,7,8,3,2,1,7,8,1,1,7,8,1,2,7,1,9,3,7,5,6,8,8,9,5,9,6,1,8,3,6,8,2,8,8,0,7,3,1,1,9,1,4,5,2,0,3,4,9,9,6,5,6,0,4,9,4,7,3,6,9,1,5,6,1,8,2,0,4,6,1,6,6,1,3,2,6,4,3,5,1,1,1,5,8,7,6,1,3,3,8,4,3,6,2,2,2,6,7,9,0,6,2,5,8,7,2,3,7,7,7,6,8,2,8,0,6,2,0,5,0,9,9,0,6,7,2,9,6,0,0,8,9,5,2,4,9,6,5,7,4,6,3,2,6,9,6,5,1,9,6,3,4,1,3,5,7,4,9,7,5,5,6,3,0,0,1,3,1,7,3,9,7,8,8,7,1,5,1,1,4,7,6,9,5,5,3,8,2,1,1,6,4,1,1,5,1,4,0,0,7,4,8,5,4,3,1,8,3,1,9,6,9,7,5,0,7,4,7,5,3,0,5,4,0,3,7,5,8,3,2,5,2,7,9,0,7,8,5,5,8,0,7,9,9,3,8,6,8,8,5,6,4,1,8,7,3,4,0,4,3,7,4,5,7,6,1,9,0,5,8,2,1,5,7,3,3,0,9,4,8,5,0,6,6,4,4,2,5,0,4,4,0,8,5,5,6,6,1,3,1,8,2,2,4,1,8,4,7,0,0,2,1,8,7,7,5,7,7,5,6,0,6,8,3,5,4,1,6,7,4,1,8,9,4,4,2,4,8,8,8,6,5,7,3,1,2,9,2,9,1,5,4,0,0,4,5,7,4,6,6,4,2,6,9,2,1,9,0,4,3,7,7,4,9,7,5,6,3,0,2,2,6,2,6,4,2,0,7,6,7,2,8,7,6,0,7,6,6,1,5,1,7,0,0,0,4,3,7,3,3,3,5,8,0,8,0,8,9,7,4,0,4,6,6,7,4,0,1,8,9,6,9,5,7,2,0,5,7,2,0,9,7,2,7,4,0,1,3,1,5,3,5,0,3,3,1,0,3,9,7,0,2,4,0,6,6,5,6,2,2,7,5,3,1,3,6,0,6,9,8,6,1,9,8,9,3,8,3,7,5,5,7,5,6,8,9,6,9,1,5,9,8,5,3,3,5,4,6,3,5,2,8,9,6,9,5,4,3,4,6,9,6,9,7,3,1,1,7,4,8,0,7,3,8,6,3,3,5,0,9,5,5,7,2,7,0,1,5,9,0,5,9,8,9,1,6,5,0,0,5,9,6,1,9,8,8,1,9,5,3,5,9,1,0,3,6,6,5,5,2,7,2,7,4,8,4,2,3,4,4,0,5,6,7,8,2,9,7,2,6,2,3,6,1,4,0,6,8,8,9,3,7,9,4,8,1,8,3,1,8,7,2,2,8,0,8,2,2,5,4,0,7,4,2,3,0,0,0,5,6,5,7,8,3,2,2,7,8,1,9,1,2,9,7,9,2,4,9,1,0,3,9,8,8,0,3,5,5,5,6,8,9,7,4,8,9,5,5,3,5,2,9,6,0,2,8,3,9,1,6,4,3,5,0,3,8,5,3,9,7,2,8,1,4,7,9,0,4,3,6,2,0,2,6,6,1,0,3,4,5,0,0,9,6,4,2,1,8,4,8,7,5,5,1,1,1,8,9,4,4,6,9,7,8,6,9,5,1,5,7,3,6,7,8,2,9,6,1,5,9,7,1,6,3,0,5,6,8,9,8,3,8,7,4,1,8,5,9,9,2,0,5,9,2,2,3,8,3,2,4,7,5,0,7,1,3,6,1,8,6,3,1,2,8,2,2,7,6,7,5,6,4,4,6,1,1,4,3,4,5,4,1,0,0,1,3,6,2,1,3,8,8,0,2,2,4,4,4,0,1,5,6,8,0,5,1,5,7,7,1,9,3,7,3,8,8,0,6,0,7,1,7,5,6,5,1,7,5,1,2,6,9,3,9,9,2,8,4,8,8,0,6,5,6,1,1,9,0,2,8,4,8,9,7,1,4,4,3,1,5,7,0,6,8,6,4,1,5,4,6,4,0,6,6,8,0,5,8,6,7,5,0,3,7,2,2,9,0,4,9,4,5,7,7,0,3,9,5,9,4,3,7,7,6,6,4,3,1,6,2,2,8,2,7,0,9,7,1,8,6,8,4,7,0,6,1,8,7,9,6,1,4,0,1,7,6,2,0,1,5,6,4,0,7,1,7,3,5,1,5,5,8,7,5,0,4,0,4,5,9,5,2,2,3,1,1,9,9,6,4,0,8,1,6,6,8,5,1,5,3,9,4,5,6,5,0,7,5,2,4,7,5,7,8,0,8,3,5,2,1,4,6,4,6,4,3,1,0,3,9,4,9,3,2,9,1,4,8,2,6,0,9,5,4,4,3,0,5,6,9,1,4,6,1,4,0,7,4,9,7,0,1,9,9,8,6,2,6,6,3,6,4,3,2,3,7,4,6,7,3,5,8,9,8,7,8,4,9,7,2,1,9,5,9,7,8,7,0,7,8,4,3,9,3,9,7,5,0,3,3,2,2,1,8,6,0,1,8,5,2,2,1,5,8,4,5,0,8,0,7,5,6,5,9,7,7,1,2,2,6,9,0,8,2,2,2,2,4,3,3,8,7,5,2,7,0,3,1,4,2,8,0,5,1,8,7,3,1,1,2,1,7,1,8,4,5,1,7,2,9,8,6,0,0,3,6,1,6,7,5,7,2,2,8,0,5,4,5,1,1,7,5,0,3,2,5,1,1,4,3,1,7,3,7,5,5,3,2,6,6,2,9,6,6,3,5,9,2,3,6,9,4,0,3,4,2,5,9,3,3,6,8,0,4,0,9,0,9,9,5,5,3,0,7,2,6,0,3,8,7,0,0,7,6,8,7,0,1,5,3,7,0,2,1,1,1,9,8,8,7,9,8,6,0,4,4,2,6,1,6,1,6,6,3,1,0,8,4,4,5,5,3,5,8,0,1,3,6,8,1,7,9,9,0,8,1,0,1,1,3,7,6,1,6,9,8,5,4,5,6,1,5,5,0,9,9,3,4,9,2,9,0,5,2,3,3,0,1,8,3,5,8,2,6,1,1,2,1,1,8,3,0,9,1,6,7,4,1,1,7,2,3,8,8,5,9,2,4,7,1,0,4,5,6,4,0,7,1,8,7,7,5,9,6,3,4,0,1,1,1,4,4,1,5,6,6,9,2,8,5,1,7,0,0,7,9,2,8,1,6,4,0,4,5,3,4,0,5,0,2,3,0,3,4,1,9,9,4,2,8,8,0,4,6,1,5,9,4,7,3,5,7,9,5,6,8,7,8,8,5,9,3,1,8,1,9,6,6,8,8,5,3,2,3,0,0,1,2,2,8,1,4,7,2,4,7,0,1,4,3,8,9,3,0,5,6,5,5,8,1,5,8,6,8,2,4,5,2,9,5,6,9,7,8,8,7,1,2,0,6,2,9,7,0,5,1,5,2,1,8,1,2,5,0,4,4,0,4,6,9,9,0,7,2,9,1,9,6,0,1,2,4,0,6,7,9,8,1,7,1,4,9,7,5,3,4,8,5,2,4,6,1,6,9,7,0,6,1,4,4,4,4,8,4,7,0,6,2,6,1,0,3,7,5,4,4,7,8,8,9,9,8,8,2,7,6,7,9,9,2,1,1,3,1,9,7,3,3,6,9,8,1,1,2,4,9,2,3,6,3,8,6,4,2,7,3,9,4,0,9,8,8,1,9,9,8,8,9,0,8,9,7,7,3,9,3,8,5,7,2,3,9,7,9,7,7,8,7,0,3,9,7,4,8,7,7,1,1,0,3,5,0,6,0,6,1,8,6,6,4,0,1,3,2,0,9,5,9,1,1,6,1,0,4,4,7,1,6,4,0,2,7,3,0,2,3,8,1,1,6,5,5,9,8,5,6,8,1,0,8,6,7,8,6,7,9,7,7,3,7,7,9,0,3,5,4,1,3,4,9,8,8,8,1,8,2,0,5,1,1,6,9,6,4,8,8,7,6,7,9,8,6,6,4,2,3,9,2,4,2,5,3,2,6,6,4,2,7,6,0,4,7,3,5,9,3,8,1,8,9,5,0,5,1,4,4,6,5,1,6,0,8,6,3,7,4,9,0,9,3,6,7,7,5,9,7,1,0,8,5,0,0,8,3,3,4,6,2,2,1,4,0,6,4,5,6,9,6,1,2,6,8,5,6,8,2,2,7,6,2,7,9,3,6,4,0,5,3,4,5,0,6,6,0,3,2,6,1,5,4,7,9,3,4,0,3,5,3,9,4,9,8,9,3,7,0,3,3,8,3,1,8,9,1,0,0,9,3,2,2,9,1,0,3,8,5,2,5,8,9,5,8,4,1,8,3,0,6,3,9,3,7,7,3,0,8,0,9,1,9,5,3,2,5,7,1,4,5,2,3,5,4,4,5,1,9,8,2,7,9,4,2,1,2,4,2,4,1,0,7,0,1,0,4,4,6,3,2,4,3,7,1,4,9,2,3,3,8,2,5,1,8,4,2,2,2,8,3,3,1,9,8,4,5,5,2,6,3,3,2,3,0,0,3,1,3,5,3,3,5,6,9,2,3,5,3,6,3,6,5,1,7,4,9,3,9,6,5,2,8,3,7,9,2,6,6,2,0,2,1,0,3,1,6,3,9,5,3,9,6,7,6,3,9,1,6,9,0,1,1,5,6,3,0,7,2,6,3,0,2,4,1,1,4,4,9,6,3,5,8,1,1,8,6,5,0,8,1,9,3,7,1,3,5,2,7,5,4,8,3,2,6,9,7,1,3,5,3,3,8,9,9,1,2,6,5,9,5,0,7,5,4,8,0,6,5,5,9,7,6,5,0,0,6,2,1,4,8,8,4,7,0,8,7,7,7,1,6,6,9,8,3,6,5,2,5,0,4,1,9,0,4,2,7,7,5,3,9,8,8,2,9,0,7,2,9,4,7,8,5,2,2,0,0,9,1,4,3,4,9,0,5,3,1,1,7,3,4,4,4,5,5,0,2,4,9,5,4,1,4,9,7,8,5,0,9,9,5,3,8,2,2,9,1,3,4,3,0,3,1,8,6,7,2,3,9,2,5,2,0,4,3,7,6,2,2,4,4,1,0,6,9,5,2,2,6,3,3,3,7,8,7,5,8,2,8,6,1,2,5,2,7,1,3,2,6,5,0,7,4,7,0,3,7,4,1,4,2,7,7,1,6,3,8,0,0,1,5,3,8,6,9,3,1,5,0,7,8,3,9,8,9,4,8,0,6,1,1,6,8,7,0,7,6,4,4,4,1,8,9,9,9,8,0,0,9,7,5,8,1,1,8,1,3,0,0,9,7,8,6,1,0,6,0,3,5,3,6,5,3,3,2,3,3,6,8,5,5,0,5,2,1,2,7,0,7,5,3,3,6,4,1,3,5,8,1,0,6,7,3,9,4,6,4,2,8,4,3,1,7,3,6,5,6,0,6,9,3,5,1,3,5,7,7,5,1,7,0,8,3,0,1,9,7,1,8,7,1,4,5,7,5,6,6,1,3,6,6,8,4,5,8,9,0,8,0,7,7,4,9,6,3,1,6,1,9,9,6,8,1,0,0,4,1,3,4,8,8,9,6,9,0,1,4,4,6,9,4,8,4,3,7,7,4,5,8,1,8,7,7,3,4,3,0,3,2,5,1,7,8,1,0,8,5,6,3,7,1,6,0,9,8,6,7,8,5,7,5,6,6,9,2,0,4,6,8,5,9,4,6,3,8,8,3,1,2,2,0,4,4,0,8,8,1,9,9,1,7,9,1,7,0,5,9,6,5,1,0,4,2,6,0,3,3,9,1,4,9,0,7,0,8,0,4,1,9,5,1,0,3,6,0,3,1,2,8,4,6,7,6,1,2,6,5,5,5,3,9,8,0,0,7,3,5,4,1,1,5,3,8,4,0,0,4,6,8,3,9,8,5,7,7,7,3,9,9,2,4,6,2,0,9,6,3,0,7,4,2,3,0,2,5,5,8,1,7,5,8,4,7,3,8,2,5,3,8,8,6,3,5,7,7,8,1,3,7,6,9,5,1,8,4,0,6,6,4,9,3,2,2,9,2,6,2,9,9,7,2,9,6,7,8,8,1,6,0,9,0,4,8,9,7,6,9,2,7,5,3,9,8,7,9,9,8,4,4,6,9,5,5,0,6,4,9,2,3,7,6,2,7,9,1,2,1,8,9,6,5,4,5,7,5,1,7,0,9,1,9,9,9,9,2,1,9,6,0,7,1,7,6,7,7,5,3,8,9,2,3,6,4,2,7,0,8,0,6,7,5,2,9,6,5,3,1,5,2,0,9,3,6,8,0,2,6,8,4,6,1,1,1,5,0,5,3,2,9,1,3,8,0,5,9,2,0,0,5,8,4,5,2,8,7,2,4,3,0,4,5,2,8,5,8,3,6,1,2,9,8,3,5,5,3,3,1,5,0,7,5,2,1,1,4,9,5,2,3,8,8,1,6,6,9,1,0,5,2,7,8,0,8,5,7,6,5,9,8,5,1,1,1,1,5,2,2,6,9,3,2,6,6,1,7,4,2,2,9,9,0,7,1,5,1,8,7,5,0,7,5,2,1,3,3,4,7,3,8,8,6,1,6,6,8,4,4,3,2,1,6,1,0,2,1,2,4,5,8,8,0,5,9,2,7,3,5,0,9,5,5,7,8,8,8,2,5,8,6,6,9,4,5,5,0,1,9,3,9,5,1,5,3,2,2,6,7,3,3,6,2,9,0,8,6,2,6,9,1,2,0,2]
    
    

# print(arr)
k = 11939
lenn = len(arr)
k = k%len(arr)
print(k,lenn)

# arr = rotateArr(arr,k)
# arr = rotateArr(arr,14)
# print(arr)



    
    
```
