```python
# 30. Write a program to find out all palindrome numbers present within a given range.
def getSign(n):
    if (n < 0):
        return -1
    return 1

def getMagnitude(n):
    if (n < 0):
        return -n
    return n

def getRev(n):
    num = getMagnitude(n)
    rev = 0
    while (num > 0):
        rev = rev*10+(num % 10)
        num = num//10

    return rev*getSign(n)

def isPalindrome(n):
    if n == getRev(n):
        return True
    return False

def getPalindromes(lb,ub):
    nums = []
    for i in range(lb,ub+1,1):
        if(isPalindrome(i)):
            nums.append(i)
    return nums

res = getPalindromes(500,700)
print(res)
```

```python
# 31. Write a program to find out all primes numbers present within a given range.



def isPrime(n):
    if(n<=1):
        return False
    half = n//2
    for i in range(2, half+1):
        if (n % i == 0):
            return False
    return True

def getPrimes(lb,ub):
    nums = []
    for i in range(lb,ub+1,1):
        if(isPrime(i)):
            nums.append(i)
    return nums

res = getPrimes(-10,100)
print(res)
```

```python
# 32. Write a program to find out all perfect square numbers present within a given range.

def getMagnitude(n):
    if (n < 0):
        return -n
    return n

def squareRoot(n):
    if n < 0:
        return 0
    if n == 0:
        return 0
    root = n / 2.0
    while 1:
        newroot = 0.5 * (root+(n/root))
        if getMagnitude(root - newroot) < 0.0000000001:
            return newroot
        root = newroot

def isPerfectSquare(n):
    if n <=0:
        return False
    root =  squareRoot(n)
    if(root==(root//1)):
        return True
    return False


def getPerfectSquares(lb,ub):
    nums = []
    for i in range(lb,ub+1,1):
        if(isPerfectSquare(i)):
            nums.append(i)
    return nums

res = getPerfectSquares(-10,100)
print(res)
```

```python
# 33. Write a program to find out all Armstrong numbers present within a given range

def getLen(n):
    n = getMagnitude(n)
    count = 0
    while (n > 0):
        count += 1
        n //= 10
    return count


def power(a,b):
    pow=1
    for i in range(b):
        pow = pow * a
    return pow


def getSumOfDigitsNPower(a,n):
    sum=0
    while(a>0):
        dig = a%10
        pow = power(dig,n)
        sum += pow
        a = a//10
    return sum
    

def isArmStrong(n):
    len = getLen(n)
    if(n-getSumOfDigitsNPower(n,len)):
        return False
    return True

def getArmStrongs(lb,ub):
    nums = []
    for i in range(lb,ub+1,1):
        if(isArmStrong(i)):
            nums.append(i)
    return nums

res = getArmStrongs(-10,10000)
print(res)
```

```python
# 34. Write a program to find out all Strong numbers present within a given range.

def factorial(n):
    if(n<=0):
        0
    fact = 1
    for i in range(1,n+1):
        fact*=i
    return fact

def getSumOfDigitsFact(n):
    sum=0
    while(n>0):
        dig = n%10
        fact = factorial(dig)
        sum += fact
        n = n//10
    return sum

def isStrong(n):
    if(n-getSumOfDigitsFact(n)):
        return False
    return True


def getStrongs(lb,ub):
    nums = []
    for i in range(lb,ub+1,1):
        if(isStrong(i)):
            nums.append(i)
    return nums

res = getStrongs(-10,100000)
print(res)

```
