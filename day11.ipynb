```python
# SORTING

# Selection sort

def selSort(arr,start,stop):
    for i in range(start,stop):
        
        for j in range(i+1,stop):
            if(arr[i]>arr[j]):
                # print(f"swaping {arr[i]} and {arr[j]}")
                temp = arr[i]
                arr[i] = arr[j]
                arr[j] = temp
    
    return arr

arr = [8,7,6,5,4,3,2,1,0]

print(arr)
n = len(arr)
selSort(arr,0,(n//2)+1)
print(arr)
selSort(arr,(n//2)+2,n)
print(arr)



```
