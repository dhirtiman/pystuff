# Even or Odd

```python

def check1(n):
    if n % 2 == 0:
        print(n, "is even")
    else:
        print(n, "is odd")


def check2(n):
    if n/2 == n//2:
        print(n, "is even")
    else:
        print(n, "is odd")


def check3(n):
    if (n/2)-(n//2) == 0:
        print(n, "is even")
    else:
        print(n, "is odd")


def check4(n):
    if (n//2)*2 == n:
        print(n, "is even")
    else:
        print(n, "is odd")


def check5(n):
    if (n//2)*2 != n:
        print(n, "is odd")
    else:
        print(n, "is even")


def check6(n):
    t = 0
    while (t < n):
        t = t+2

    if (t == n):
        print(n, "is even")
    else:
        print(n, "is odd")


def check7(n):
    t = n
    while (t > 0):
        t = t-2

    if (t == 0):
        print(n, "is even")
    else:
        print(n, "is odd")

                      
def check6(n):
    t=0
    while (t<n):
        t=t+2
    
    if (t==n):
        print(n,"is even")
    else:
        print(n,"is odd")


                      
def check8(n):
    ld = n%10
    if(ld == 0 or ld == 2 or ld == 4 or ld == 8):
        print(n,"is even")
    else:
        print(n,"is odd")
        
        
def check9(n):
    if( (n^1)>n):
        print(n,"is even")
    else:
        print(n,"is odd")
        

def check10(n):
    if(n&1): #  even&1 = 0 , odd&1 = 1
        print(n,"is odd")
    else:
        print(n,"is even")
              
def check11(n):
    if ((n*5)%10):
        print(n, "is odd")
    else:
        print(n, "is even")
        


def check12(n):
    if (n >> 1 << 1 == n):
        print(n, "is even")
    else:
        print(n, "is odd")
        
        

def check13(n):
    if (n // 2) << 1 == n :
        print(n, "is even")
    else:
        print(n, "is odd")
        

def check14(n):
    if (n | 1) == n+1 :
        print(n, "is even")
    else:
        print(n, "is odd")        

def check15(n):
    if (n | 1) ==  n :
        print(n, "is odd")     
    else:
        print(n, "is even")

        
def check16(n):
    if (n^1)+1==n:
        print(n,"is odd")
    else:
        print(n,"is even")


check1(2)
check1(3)
check3(4)
check4(5)
check5(6)
check6(7)
check7(8)
check8(934)
check9(9)
check10(10)
check11(11)
check12(12)
check13(13)
check14(14)
check15(15)
check16(16)
```
