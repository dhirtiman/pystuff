```python
#8
def extract2ndHalf(num):
    len = 0
    temp = num
    while (temp > 0):
        len += 1
        temp = temp//10
    half = len//2

    n = num
    tens = 1
    while (half > 0):
        n = n//10
        tens = tens*10
        half -= 1

    num = num % tens
    return num

print(extract2ndHalf(11))
print(extract2ndHalf(123))
print(extract2ndHalf(1234))
print(extract2ndHalf(12345))
print(extract2ndHalf(5464))
```
