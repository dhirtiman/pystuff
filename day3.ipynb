```python

def greet(time):
    # time = float(input("enter time in 24-hour format (hh.mm)"))
    minute = (time *100)%100
    if ( time < 0 or time > 23.59 or minute > 59):
        print("Invalid Time")
    elif (time <= 11.59):
        print("Good Morning")

    elif (time <= 15.59):
        print("Good Afternoon")

    elif (time <= 18.59):
        print("Good Evening")
    else:
        print("Good Night")


greet(9)
greet(10.50)
greet(13.5)
greet(16.1)
greet(23.59)
greet(24)
greet(14.60)



```
