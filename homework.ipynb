```python
def tri1(n):
    for i in range(n):
            
        for j in range(n):
            if(j==0 or i == n-1 or (i-j)==0):
                print("*",end=" ")
            else:
                print(" ",end=" ")
                
        print()

tri1(5)
```

```python
def tri2(n):
    for i in range(n):
            
        for j in range(n):
            if(j==0 or i == 0 or i+j == n-1):
                print("*",end=" ")
            else:
                print(" ",end=" ")
                
        print()

tri2(5)
```
