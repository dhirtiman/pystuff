```python
def saystuff(people):
    for person in people:
        print("My name is", person["name"], "Im ", person["age"])


people = [
    {
        "name": "Dhirtiman",
        "age": 22
    },
    {
        "name": "Bikash Yadav",
        "age": 30
    },
    {
        "name": "A Aswani Kumar",
        "age": 22
    },
    {
        "name": "Ekwa Bebey Christian",
        "age": 38
    },
    {
        "name": "Shivanand bro",
        "age": 23
    }
]

saystuff(people)
```
