```python
# 3. Write a program to compute x^n/n!.


def power(a,b):
    pow=1
    for i in range(b):
        pow = pow * a
    return pow

def factorial(n):
    fact = 1
    for i in range(1,n+1):
        fact*=i
    return fact


def getXpNByNFact(x,n):
    return (power(x,n))/factorial(n)

res = getXpNByNFact(2,2)
print(res)

res = getXpNByNFact(2,3)
print(res)
```

```python
# 4. Write a program to count the number of digits in an integer.


def getMagnitude(n):
    if (n < 0):
        return -n
    return n


def getDigitCount(n):
    n = getMagnitude(n)
    count = 0
    while (n > 0):
        count += 1
        n //= 10
    return count


def getLen(num):
    return getDigitCount(num)


print(getDigitCount(-1232))
print(getDigitCount(124314))
print(getLen(12345))
```

```python
# 5. Write a program to sum all digits of a number.

def getDigitSum(n):
    n = getMagnitude(n)
    len = getLen(n)
    sum = 0
    for i in range(len):
        dig = n % 10
        sum += dig
        n = n//10
    return sum

print(getDigitSum(-123))
print(getDigitSum(123))
print(getDigitSum(12345))
```

```python
# 6. Write a program to reverse a given number.

def getSign(n):
    if (n < 0):
        return -1
    return 1


def getRev(n):
    num = getMagnitude(n)
    rev = 0
    while (num > 0):
        rev = rev*10+(num % 10)
        num = num//10

    return rev*getSign(n)


print(getRev(54321))
print(getRev(-123))
print(getRev(-54321))
```

```python
def getGCD(a,b):
    while(b>0):
        rem = a%b
        a = b
        b = rem
    return a    
    
def getHCF(a,b):
    return getGCD(a,b)
    
print(getGCD(12, 8))     # Expected: 4
print(getGCD(100, 25))   # Expected: 25
print(getGCD(17, 13))    # Expected: 1 (both prime)
print(getGCD(36, 60))    # Expected: 12
print(getGCD(81, 27))    # Expected: 27
print(getGCD(7, 0))      # Expected: 7
print(getHCF(0, 9))      # Expected: 9
print(getHCF(0, 0))      # Expected: 0 
print(getHCF(1, 999))    # Expected: 1
print(getHCF(54, 24))    # Expected: 6
```

```python
# 8. Write a program to calculate the LCM of two numbers.

# LCM(a,b) = |a x b| / gcd(a,b) 

def getLCM(a, b):
    if (a == 0 and b == 0):
        return 0
    return getMagnitude(a*b)//getGCD(a, b)


print(getLCM(12, 18))
print(getLCM(4, 6))       # Expected: 12
print(getLCM(10, 5))      # Expected: 10
print(getLCM(7, 3))       # Expected: 21
print(getLCM(12, 18))     # Expected: 36
print(getLCM(8, 16))      # Expected: 16
print(getLCM(1, 9))       # Expected: 9
print(getLCM(9, 1))       # Expected: 9
print(getLCM(0, 5))       # Expected: 0
print(getLCM(5, 0))       # Expected: 0
print(getLCM(0, 0))       # Expected: 0
print(getLCM(13, 13))     # Expected: 13
print(getLCM(21, 6))      # Expected: 42
```

```python
# 10. Write a Program to calculate prime Factors of number.


def isPrime(n):
    if(n<=0):
        return False
    half = n//2
    for i in range(2, half+1):
        if (n % i == 0):
            return False
    return True

def getFactors(n):
    factors = {1,n}
    for i in range(2,n):
        if(n%i==0):
            factors.add(i)
    return factors



def getNPrimes(n):
    primes = []
    count = n
    i=1
    while(count>0):
        if(isPrime(i)):
            primes.append(i)
            count-=1
        i+=1
    return primes
    
def getPrimeFactors(n):
    
    factors = []
    for i in range(2,n+1):
        if(n%i==0):
            factors.append(i)
            while(n%i==0):
                n = n//i
    
    return factors
    
print(getPrimeFactors(2))     # [2]
print(getPrimeFactors(3))     # [3]
print(getPrimeFactors(4))     # [ 2]
print(getPrimeFactors(6))     # [2, 3]
print(getPrimeFactors(8))     # [2]
print(getPrimeFactors(12))    # [ 2, 3]
print(getPrimeFactors(18))    # [2, 3]
print(getPrimeFactors(25))    # [5]
print(getPrimeFactors(49))    # [7]
print(getPrimeFactors(60))    # [2, 3, 5]
print(getPrimeFactors(97))    # [97]  # prime
print(getPrimeFactors(1))     # [] 
print(getPrimeFactors(81))     # [3] 
```

```python
# 11. Write a program to check the given number is a palindrome or not.


def getSign(n):
    if (n < 0):
        return -1
    return 1

def getMagnitude(n):
    if (n < 0):
        return -n
    return n

def getRev(n):
    num = getMagnitude(n)
    rev = 0
    while (num > 0):
        rev = rev*10+(num % 10)
        num = num//10

    return rev*getSign(n)

def isPalindrome(n):
    if n == getRev(n):
        return True
    return False


print(isPalindrome(121111121))  # True
print(isPalindrome(121))        # True
print(isPalindrome(123))        # False
print(isPalindrome(0))          # True
print(isPalindrome(7))          # True
print(isPalindrome(1221))       # True
print(isPalindrome(12321))      # True
print(isPalindrome(12345))      # False
```

```python
# 12. Write a program to check whether a given number is prime or not.


def isPrime(n):
    if(n<=0):
        return False
    half = n//2
    for i in range(2, half+1):
        if (n % i == 0):
            return False
    return True


print(isPrime(2))     # True
print(isPrime(3))     # True
print(isPrime(4))     # False
print(isPrime(5))     # True
print(isPrime(1))     # False
print(isPrime(0))     # False
print(isPrime(-7))    # False
print(isPrime(9))     # False
print(isPrime(11))    # True
print(isPrime(15))    # False
print(isPrime(17))    # True
print(isPrime(97))    # True
```

```python
# 13. Write a program to check whether a given number is a perfect square number or not.

# Newton-Raphson method
# xnew = 0.5 x (xold + n/xold)

def getMagnitude(n):
    if (n < 0):
        return -n
    return n

def squareRoot(n):
    if n < 0:
        return 0
    if n == 0:
        return 0
    root = n / 2.0
    while 1:
        newroot = 0.5 * (root+(n/root))
        if getMagnitude(root - newroot) < 0.000001:
            return newroot
        root = newroot

def isPerfectSquare(n):
    root =  squareRoot(n)
    if(root==(root//1)):
        return True
    return False

print(isPerfectSquare(25)) # True
print(isPerfectSquare(50)) # False
print(isPerfectSquare(625)) # True
print(isPerfectSquare(1000)) # False

```
