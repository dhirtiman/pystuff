```python
# 13. Write a program to check whether a given number is a perfect square number or not.

# Newton-Raphson method
# xnew = 0.5 x (xold + n/xold)

def getMagnitude(n):
    if (n < 0):
        return -n
    return n

def squareRoot(n):
    if n < 0:
        return 0
    if n == 0:
        return 0
    root = n / 2.0
    while 1:
        newroot = 0.5 * (root+(n/root))
        if getMagnitude(root - newroot) < 0.0000000001:
            return newroot
        root = newroot

def isPerfectSquare(n):
    if(n<=0):
        return False
    root =  squareRoot(n)
    if(root==(root//1)):
        return True
    return False

print(isPerfectSquare(25)) # True
print(isPerfectSquare(16)) # True
print(isPerfectSquare(50)) # False
print(isPerfectSquare(625)) # True
print(isPerfectSquare(1000)) # False

```

```python
# 19. Write a program to check whether a given number is an Automorphic number or not.
# An automorphic number is a number whose square ends with the original number itself

def getTensOfNum(n):
    tens=1
    while(n>0):
        tens=tens*10
        n = n //10
    return tens

def power(a,b):
    pow=1
    for i in range(b):
        pow = pow * a
    return pow

def isAutomorphic(n):
    if (n<=0):
        return False
    square = power(n,2)
    tens = getTensOfNum(n)    
    lpart = square%tens
    if(n-lpart==0):
        return True
    return False



print(isAutomorphic(5))
print(isAutomorphic(76))
print(isAutomorphic(6))
print(isAutomorphic(376))

```

```python
# 20. Write a Program to check whether the number is Magic Number or Not
# A number is a Magic Number if the sum of its digits is repeatedly calculated
# until a single digit is obtained, and that digit is 1.


def getSumDigits(n):
    sum = 0
    while (n>0):
        dig = n % 10
        sum += dig
        n = n//10
    return sum

getSumDigits(127)
getSumDigits(10)

def isMagic(n):    
    while(n>9):
        n = getSumDigits(n)
    if(n==1):
        return True
    return False
    

print(isMagic(127))
print(isMagic(19))


for i in range(100):
    if(isMagic(i)):
        print(i, end=" ")
```

```python
# 21. Write a program to check whether a given number is Friendly pair or not.

# A Friendly Pair (or friendly numbers) are two natural numbers 
# whose theratio of the sum of their proper divisors to the number itself—is equal.


def getSumOfFactors(n):
    sum = 0
    for i in range(1,n+1):
        if(n%i==0):
            sum+=i
    return sum

getSumOfFactors(28)
getSumOfFactors(6)

def areFriendly(a,b):
    suma = getSumOfFactors(a)
    sumb = getSumOfFactors(b)
    ratioa  = suma/a
    ratiob = sumb/b
    if(ratioa==ratiob):
        return True
    return False

print(areFriendly(6,28))
print(areFriendly(30, 140))
print(areFriendly(80, 200))
print(areFriendly(40, 224))




```

```python
# 24. Write a Program to check whether the number is Happy Number or Not.
# A happy number is a positive integer that eventually reaches 1
# when replaced by the sum of the squares of its digits repeatedly.
# If this iterative process enters an infinite cycle that does not include 1, it is considered an unhappy (or sad) number

def getSumOfDigitsNPower(a, n):
    sum = 0
    while (a > 0):
        dig = a % 10
        pow = power(dig, n)
        sum += pow
        a = a//10
    return sum


def isHappy(n):
    seen = set()
    while 1:
        n = getSumOfDigitsNPower(n, 2)
        if (n == 1):
            return True
        if (n in seen):
            return False
        seen.add(n)
    return False

print(isHappy(7))
print(isHappy(23))
print(isHappy(97))
print(isHappy(68))


for i in range(1, 100+1):
    if (isHappy(i)):
        print(i, end=" ")
```

```python
# 25. Write a Program to check whether the number is Sunny Number or Not.
# A sunny number is a positive integer n where its immediate successor, n+1 is a perfect square


def squareRoot(n):
    if n < 0:
        return 0
    if n == 0:
        return 0
    root = n / 2.0
    while 1:
        newroot = 0.5 * (root+(n/root))
        if getMagnitude(root - newroot) < 0.0000000001:
            return newroot
        root = newroot

def isPerfectSquare(n):
    if(n<=0):
        return False
    root =  squareRoot(n)
    if(root==(root//1)):
        return True
    return False

def isSunny(n):
    if(n<=0):
        return False
    return isPerfectSquare(n+1)

print(isSunny(3))
print(isSunny(8))
print(isSunny(15))
print(isSunny(24))
```

```python
# 26. Write a Program to check whether the number is Disarium Number or Not.
# A Disarium number is a number where the sum of its digits raised to the power of their respective positions 
# (starting from 1) equals the number itself


def getSumOfDigitsPositionPower(n):
    sum = 0
    pos = getLen(n)
    while(n>0):
        dig = n%10
        n = n // 10
        pow = power(dig,pos)
        sum += pow
        pos-=1
    return sum

def isDisarium(n):
    if (n == getSumOfDigitsPositionPower(n)):
        return True
    return False


print(getSumOfDigitsPositionPower(89))
print(isDisarium(89))
print(isDisarium(135))
print(isDisarium(175))

```

```python
# 27. Write a Program to check whether the number is a Pronic Number or Not.

# A Pronic number is any integer that's the product of two consecutive integers

# for i in range (100):
#     print(i,i+1,i*(i+1))

def squareRoot(n):
    if n < 0:
        return 0
    if n == 0:
        return 0
    root = n / 2.0
    while 1:
        newroot = 0.5 * (root+(n/root))
        if getMagnitude(root - newroot) < 0.0000000001:
            return newroot
        root = newroot
        

def isPronic(n):
    root = squareRoot(n)
    root = root//1
    prod = root*(root+1)
    if(prod==n):
        return True
    return False


print(isPronic(110))
print(isPronic(20))
print(isPronic(30))
print(isPronic(12))

```

```python
# 28. Write a Program to check whether the number is a Trimorphic Number or Not.


def getTensOfNum(n):
    tens=1
    while(n>0):
        tens=tens*10
        n = n //10
    return tens

def power(a,b):
    pow=1
    for i in range(b):
        pow = pow * a
    return pow

def isTrimorphic(n):
    square = power(n,3)
    tens = getTensOfNum(n)    
    lpart = square%tens
    if(n-lpart==0):
        return True
    return False


print(isTrimorphic(5))
print(isTrimorphic(24))
print(isTrimorphic(249))
```
