```python
def isPerfectSquare( num):
        """
        :type num: int
        :rtype: bool
        """
        root = num /2.0
        while 1:
            newroot = 0.5 * (root+(num/root))
            if(abs(root - newroot)<0.0000000001):
                root = newroot
                break
            root = newroot
        if(root==int(root)):
            return True
        return False
    

print(isPerfectSquare(16))
```

```python
def getSumOfDigitsNPower(a, n):
    sum = 0
    while (a > 0):
        dig = a % 10
        pow = dig**n
        sum += pow
        a = a//10
    return sum


def isHappy(n):
    seen = set()
    while 1:
        n = getSumOfDigitsNPower(n, 2)
        if (n == 1):
            return True
        if (n in seen):
            return False
        seen.add(n)
    return False

for i in range(100):
    if(isHappy(i)):
        print(i,end=" ")
```
