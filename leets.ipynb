```python
# 33. Search in rotated sorted array
# Input: 
# Output: 4





from typing import List


def search(nums: List[int],target: int) -> int:
    n: int = len(nums)
    i: int = 0
    j: int = n-1
    
    while i<=j:
        mid = (i+j)//2
        if(nums[mid]==target):
            return mid
        if(nums[i]<=nums[mid]):
            if(nums[i]<=target<=nums[mid]):
                j = mid -1
            else:
                i = mid +1
        elif(nums[mid]<=target<=nums[j]):
            i = mid+1
        else:
            j = mid-1
    
    return -1



nums = [4,5,6,7,0,1,2]
target = 0
res = search(nums,target)
print(res)
```

```python
def isPrime(n):
    if (n < 2):
        return False
    if (n < 4):
        return True
    if (n % 2 == 0 or n % 3 == 0):
        return False
    i = 5
    while i*i <= n:
        if (n % i == 0 or n % (i+2) == 0):
            return False
        i += 6
    return True


def hasPrimeSetBits(n):
    count = 0
    while (n > 0):
        if (n % 2):
            count += 1
        n = n//2
    if (isPrime(count)):
        return count
    return -1


countSet = set()
for i in range(1, (10**6)+1):
    countSet.add(hasPrimeSetBits(i))
    
    
print(list(countSet))

[2, 3, 5, 7, 11, 13, 17, 19]
```

```python
def getAllCombos(char_set, prefix, n, k,combos):
    if k == 0:
        combos.append(prefix)
        return
    for i in range(n):
        new_prefix = prefix + char_set[i]
        getAllCombos(char_set, new_prefix, n, k - 1,combos)

def getAllComb(char1, char2, N):
    char_set = [char1, char2]
    # Call the recursive function
    combos = []
    getAllCombos(char_set, "", len(char_set), N,combos)
    return combos

# Example usage:
# combos = getAllComb('0', '1', 7)

    
```

```python
def isDeciBin(n):    
    while(n>0):
        dig=n%10
        if(not (dig==0 or dig ==1)):
            return False
        n//=10
    return True

def getAllDeciBinNums(k,n):
    deciBins = []
    for i in range(k,n+1):
        if(isDeciBin(i)):
            deciBins.append(i)
    
    return deciBins

a = 3
tens = ((10**a)-1)-(10**(a-1)-1)
thing = (10**a)
deciBins = getAllDeciBinNums(1,200)
stuff = (thing-tens)+()
print(deciBins)
print(tens,thing,stuff)


```
