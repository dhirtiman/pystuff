```python
def findPeakElement(nums):
    n = len(nums)
    if(n==1):
        return 0
    i = 1
    
    while 1:
        if(nums[i]>=nums[(i+1)%n]):
            if(nums[i]>=nums[(i-1)%n]):
                return i
            else:
                i = (i -1)%n
        elif(nums[(i+1)%n]>=nums[(i-1)%n]):
            i = (i+1)%n
        else:
            i = (i-1)%n
    
        
nums = [1,2,3,1]
nums = [1]


nums = [2,1,2]
nums = [5,4,3,4,5]
nums = [-7,-8,-19,24,19,3]

res = findPeakElement(nums)
print(res)
# i = 1
# print(nums[i]>=nums[(i+1)%6])
# print(nums[i]>=nums[(i-1)%6])
# print((i-1)%6)
```
